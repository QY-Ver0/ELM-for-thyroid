```python
def corr_df(col_name, target, percentage=False):
    df_freq = None
    df_percent = None
    for stage in df[col_name].unique():
        df_temp = (df.loc[df[col_name]==stage, [col_name, target]]
                   .value_counts() # Get frequency
                   .to_frame()    # Series -> DataFrame
                   .reset_index() # Was multi-index
                   .pivot(index=col_name, columns=target, values='count')) # Set index to recurred
        # print(df_temp.sum(axis=1))
        freq = df_temp.sum(axis=1)[stage]

        df_temp2 = df_temp.divide(freq)
        if percentage and df_percent is None:
            df_percent = df_temp2
        else:
            df_percent = pd.concat([df_percent, df_temp2], axis=0)

        if not percentage and df_freq is None:
            df_freq = df_temp
        else:
            df_freq = pd.concat([df_freq, df_temp], axis=0)

    df_freq.fillna(0, inplace=True)
    df_percent.fillna(0, inplace=True)
    if percentage:
        return df_percent
    else:
        return df_freq

# Stage vs Recurred
df_stage_percent = corr_df('Stage', 'Recurred', percentage=True)
df_stage_freq = corr_df('Stage', 'Recurred', percentage=False)
df_stage_freq = df_stage_freq.reindex(['I', 'II', 'III', 'IVA', 'IVB'], axis=0)
df_stage_percent = df_stage_percent.reindex(['I', 'II', 'III', 'IVA', 'IVB'], axis=0)
df_stage_freq.plot(kind='bar')
plt.title('Recurrence based on Stage')
df_stage_percent.plot(kind='bar', stacked=True)
plt.title('Recurrence based on Stage - Percentage')

df_risk_percent = corr_df('Risk', 'Recurred', percentage=True)
df_risk_freq = corr_df('Risk', 'Recurred', percentage=False)
df_risk_freq.plot(kind='bar')
plt.title('Recurrence based on Risk')
df_risk_percent.plot(kind='bar', stacked=True)
plt.title('Recurrence based on Risk - Percentage')

df_response_percent = corr_df('Response', 'Recurred', percentage=True)
df_response_freq = corr_df('Response', 'Recurred', percentage=False)
df_response_freq = df_response_freq.reindex(['Indeterminate', 'Structural Incomplete', 'Biochemical Incomplete', 'Excellent'], axis=0)
df_response_percent = df_response_percent.reindex(['Indeterminate', 'Structural Incomplete', 'Biochemical Incomplete', 'Excellent'], axis=0)
df_response_freq.plot(kind='bar')
plt.title('Recurrence based on Response')
df_response_percent.plot(kind='bar', stacked=True)
plt.title('Recurrence based on Response - Percentage')

plt.show()
```

```python
cat_risk = [['Low', 'Intermediate', 'High']]
cat_T = [['T1a', 'T1b', 'T2', 'T3a', 'T3b', 'T4a', 'T4b']]
cat_N = [['N0', 'N1a', 'N1b']]
cat_M = [['M0', 'M1']]
cat_stage = [['I', 'II', 'III', 'IVA', 'IVB']]
onehot_cols1 = ['Gender', 'Smoking', 'Hx Smoking', 'Hx Radiotherapy', 'Thyroid Function', 'Physical Examination', 'Adenopathy', 'Pathology', 'Focality']
onehot_cols2 = ['Response', 'Recurred']

# Firstly, change the name of this column, authors' grammar mistake
df.rename(columns={'Hx Radiothreapy': 'Hx Radiotherapy'}, inplace=True)
preprocessor = ColumnTransformer(
    transformers=[
        ('minmax_age', MinMaxScaler(), ['Age']),
        ('onehot1', OneHotEncoder(categories='auto', drop='if_binary', dtype=bool, sparse_output=False), onehot_cols1),
        ('encode_risk', OrdinalEncoder(categories=cat_risk), ['Risk']),
        ('encode_T', OrdinalEncoder(categories=cat_T), ['T']),
        ('encode_N', OrdinalEncoder(categories=cat_N), ['N']),
        ('encode_M', OrdinalEncoder(categories=cat_M), ['M']),
        ('encode_stage', OrdinalEncoder(categories=cat_stage), ['Stage']),
        ('onehot2', OneHotEncoder(categories='auto', drop='if_binary', dtype=bool, sparse_output=False), onehot_cols2),
    ],
    remainder='passthrough',
    force_int_remainder_cols=False,
    verbose_feature_names_out=False,
    sparse_threshold=0
)
preprocessor.fit(df)
# preprocessor.get_feature_names_out()
```

```python
# transform preprocessor
df_preprocessed = pd.DataFrame(preprocessor.transform(df), columns=preprocessor.get_feature_names_out())
# to int
# df_preprocessed.dtypes
```
