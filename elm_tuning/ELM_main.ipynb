```python
import numpy as np
import pandas as pd

# scikit-learn == 1.5.2
from sklearn.model_selection import train_test_split
import time

from ELM import ELMclf
from preprocess import preprocess, get_data

from utils import cross_val_score, get_metrics
```

```python
split_seed = 42
```

```python
path_to_data = '../data/'
```

```python
# Acquire preprocessed dataset
df = get_data(f'{path_to_data}thyroid_diff.csv')
df = preprocess(df, do_onehot=True, do_scale=True)

X = df.drop('Recurred_Yes', axis=1)
Y = df['Recurred_Yes']
x_train,     x_test,     y_train,     y_test     = train_test_split(X, Y, test_size=0.2, random_state=split_seed, stratify=Y)
x_train_sub, x_validate, y_train_sub, y_validate = train_test_split(x_train, y_train, test_size=0.25, random_state=split_seed, stratify=y_train)
```

```python
###
### Only Neuron cnt
def multimodel_cross_val_score(
        estimator, X_param, y=None,
        scoring=None, cv=None, shuffle=False, stratified=True,
        random_seed_range=range(0, 100)):
    ls = []
    for i_inner in random_seed_range:
        estimator.random_state = i_inner
        score_i = cross_val_score(estimator, X_param, y, scoring, cv, shuffle, stratified, random_state=i_inner)
        ls.append(score_i)

    return np.array(ls)


def generate_with_parameters(neurons_cnt=range(1,150), l2_params=np.arange(0.,1.), model_seeds=range(42,43)):
    # texts = []
    neuron_for_pd     = []
    l2_for_pd         = []
    model_seed_for_pd = []

    all_metric_names = ['acc', 'prec', 'rec', 'f1', 'f2', 'spec', 'npv', 'mcc', 'generic']
    all_types = ['avgs', 'maxs', 'mins', 'stds']
    all_metrics_dict = {f'{m}_{stat_type}':[] for m in all_metric_names for stat_type in all_types}

    scoring_func = get_metrics
    for i_inner in neurons_cnt:
        start = time.time()
        for l2 in l2_params:
            l2_start = time.time()
            ls = np.array(multimodel_cross_val_score(ELMclf(i_inner, l2_param=l2), x_train.to_numpy(), y_train.to_numpy(), cv=5, scoring=scoring_func, random_seed_range=model_seeds))

            # if l2 == 0: texts.append(f' ELM of {i_inner} neurons')
            # else:       texts.append(f'RELM of {i_inner} neurons & l2_param of {l2}')

            # if not get metrics, just return one
            if scoring_func.__name__ != 'get_metrics':
                for _, _ in enumerate(model_seeds):
                    neuron_for_pd.append(i_inner)
                    l2_for_pd.append(l2)

                    all_metrics_dict['generic_avgs'].append(ls.mean())
                    all_metrics_dict['generic_maxs'].append(ls.max())
                    all_metrics_dict['generic_mins'].append(ls.min())
                    all_metrics_dict['generic_stds'].append(ls.std())
            else:
                # decompose the list of dicts
                for idx, m in enumerate(model_seeds):
                    neuron_for_pd    .append(i_inner)
                    l2_for_pd        .append(l2)
                    model_seed_for_pd.append(m)

                    metric_full_names = ['accuracy', 'precision', 'recall', 'f1', 'f2', 'specificity', 'npv', 'mcc']
                    inner_dict = {m : np.array([metrics[full_m] for metrics in ls[idx]])
                                  for m, full_m in zip(all_metric_names,metric_full_names)}

                    for m, inner_ls in inner_dict.items():
                        all_metrics_dict[f'{m}_avgs'].append(inner_ls.mean())
                        all_metrics_dict[f'{m}_maxs'].append(inner_ls.max())
                        all_metrics_dict[f'{m}_mins'].append(inner_ls.min())
                        all_metrics_dict[f'{m}_stds'].append(inner_ls.std())

            print(f"{time.time() - l2_start:.4f} s: Neuron {i_inner}, L2 {l2}")
        print(f"\t\t\tNeuron {i_inner}: {time.time() - start:.2f} seconds")

    if scoring_func.__name__ != 'get_metrics':
        df_param = pd.DataFrame(np.transpose(np.array(
            [
             neuron_for_pd,
             l2_for_pd,
             np.array(all_metrics_dict['generic_avgs']),
             np.array(all_metrics_dict['generic_maxs']),
             np.array(all_metrics_dict['generic_mins']),
             np.array(all_metrics_dict['generic_stds']),]
        )), columns=['NeuronsCnt', 'L2', 'Avg', 'Max', 'Min', 'Std'])
    else:
        list_for_df = (
                [model_seed_for_pd, neuron_for_pd, l2_for_pd] +
                [np.array(all_metrics_dict[f'{m}_{stat_type}']) for m in all_metric_names[:-1] for stat_type in all_types])
        col_names = (
                ['ModelSeed', 'NeuronsCnt', 'L2'] +
                [f'{t.capitalize()}_{stat_type.capitalize()[:-1]}'
                 for t in all_metric_names[:-1] for stat_type in all_types])
        df_param = pd.DataFrame(np.transpose(np.array(list_for_df)), columns=col_names)
    return df_param
```

```python
n_low, n_high = 2, 100
neurons = range(n_low, n_high+1)

l2_low, l2_high, l2_inc = 0., 1., 1
l2s = np.arange(l2_low, l2_high, l2_inc)

seed_low, seed_high, seed_inc = 0, 50, 2
seeds = range(seed_low, seed_high, seed_inc)
hidden_neuron_df = generate_with_parameters(neurons, l2s, seeds)
```

```python
fname = f'{path_to_data}node{n_low}_{n_high}__seed{seed_low}_{seed_high}_{seed_inc}__l0_1_{str(l2_inc).replace('.','')}.csv'
hidden_neuron_df.to_csv(fname)
```
