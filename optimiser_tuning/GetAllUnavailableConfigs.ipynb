```python
import json
import os
```

```python
def get_missing_params(s_list, mi_list, mult, employed_legacy=True, onlooker_legacy=True, ver=None):
    # btw cv_on_train doesn't affect anything anymore, now ABC disregards the validation passed in entirely
    all_case = [{
        'solution_cnt': solu_cnt,
        'max_iter': max_i,
        'trial_limit': solu_cnt * mult,
    } for solu_cnt in s_list for max_i in mi_list]

    # ver -1 for getting all case
    if ver == -1:
        return map(lambda x: (x, False), all_case)

    return zip(all_case, [os.path.exists(f'../data/abc_scores_final_n{int(employed_legacy)}{int(onlooker_legacy)}_sol{solu_cnt}_iter{max_i}_lim{solu_cnt*mult}.csv') for solu_cnt in s_list for max_i in mi_list])
```

```python
# specify the grid of parameters required here
all_s      = [15, 20, 30, 45, 60, 75]
all_mi     = [200,300,500,750,1000]
employed_legacy = True
onlooker_legacy = True
cv_on_train     = False
ver = 1

path_exists = list(get_missing_params(all_s,all_mi, 2, employed_legacy, onlooker_legacy, ver))
not_yet_available = [tup for tup in path_exists if not tup[1]]
available = [tup for tup in path_exists if tup[1]]
len(not_yet_available)
```

```python
# Throw them into a JSON
not_available_list = [tup[0] for tup in not_yet_available]
with open('not_yet_available.json', 'w') as f:
    json.dump(not_available_list, f, indent=4)
```
