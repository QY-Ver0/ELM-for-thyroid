```python
import json
import nbformat as nbf
```

```python
with open('not_yet_available.json', 'r') as file:
    not_yet_available = json.load(file)
```

```python
param_list = not_yet_available
for d in param_list:
    d['employed_legacy'] = True
    d['onlooker_legacy'] = True
```

```python
len(param_list)
```

```python
path = '../optimisation_executions'
```

```python
# creates a Notebook for each parameter required,
# then either run each iteratively, or distribute it to multiple device to run all in Parallel
# (given that it is possible to have that many device)
def create_execution_notebook(abc_param, p):
    param_string = f'{abc_param}'
    fname = f'{p}/OptimiserExec_final_n{int(abc_param['employed_legacy'])}{int(abc_param['onlooker_legacy'])}_s{abc_param['solution_cnt']}_m{abc_param['max_iter']}_l{abc_param['trial_limit']}.ipynb'
    cell1 = '''from optimisation_exec_ops.OptimisationParamExecution import OptimisationParamExecution'''
    cell2 = f'''param = {{
    'n'             : 38,
    'l2'            : 0.36,
    'def_rng_seed'  : 42,
    'split_seed'    : 42,
    'cv'            : 5,
    'ver'           : 2, #this is for final
    'path_directory': '../data',
}}
abc_param = {param_string}
'''
    cell3 =  '''execution = OptimisationParamExecution(**param)
execution.fit(**abc_param)'''

    nb = nbf.v4.new_notebook()
    nb['cells'] = [
        nbf.v4.new_code_cell(cell1),
        nbf.v4.new_code_cell(cell2),
        nbf.v4.new_code_cell(cell3)
    ]

    with open(fname, "w") as f:
        nbf.write(nb, f)
    print(f'File with name {fname} created.')
```

```python
for i, param in enumerate(param_list):
    print(f'{i}: ', end='')
    create_execution_notebook(param, path)
```
