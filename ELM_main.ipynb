```python
def multimodel_cross_val_score(estimator, X, y=None, scoring=None, cv=None, shuffle=False, stratified=True, random_seed_range=range(0,100)):
    ls = []
    for i in random_seed_range:
        estimator.random_state = i
        score_i = cross_val_score(estimator, X, y, scoring, cv, shuffle, stratified, random_state=i)
        ls.append(score_i)

    # print(ls)
    return np.array(ls)
```

```python
# custom function only for
def specificity_score(yt, yp, zero_division=0):
    tn, fp, fn, tp = confusion_matrix(yt, yp).ravel().tolist()
    if tn+fp == 0:
        return zero_division
    return tn/(tn+fp)
```

```python
def plot_with(df, title=None, use_scatter=True):
    """
    Given dataframe with 5 columns, plot them
    :param df:
    :param title:
    :param use_scatter:
    :return:
    """
    x_axis, col_avg, col_max, col_min, col_std = df.columns
    if use_scatter:
        axs = df.plot.scatter(x_axis, col_avg, color='white', title=title)
        df.plot.scatter(x_axis, col_max, color='cyan', ax=axs)
        df.plot.scatter(x_axis, col_min, color='orange', ax=axs)
    else:
        axs = df.plot.line(x_axis, col_avg, color='white', title=title)
        df.plot.line(x_axis, col_max, color='cyan', ax=axs)
        df.plot.line(x_axis, col_min, color='orange', ax=axs)
    # df.plot.line(x_axis, col_std, color='green', ax=axs)
    plt.fill_between(df[x_axis], df[col_avg] - df[col_std], df[col_avg] + df[col_std], alpha=0.3, label='Std') # std area
    # df.plot.line('NeuronsCnt', 'Avg/Std', color='green')
    # axs.plot(df['NeuronsCnt'], df['Max']-df['Min'], color='green')


    print(f'Best {col_max} performance: ')
    max_best = df.loc[df[col_max].idxmax()]
    print(max_best)
    plt.axvline(max_best[x_axis], color='cyan', linestyle='dotted', label='Best Max', alpha=0.8)

    print(f'Best {col_min} performance: ')
    min_best = df.loc[df[col_min].idxmax()]
    print(min_best)
    plt.axvline(min_best[x_axis], color='orange', linestyle='dashed', label='Best Min', alpha=0.5)

    print(f'Best {col_avg} Performance: ')
    avg_best = df.loc[df[col_avg].idxmax()]
    print(avg_best)
    plt.axvline(avg_best[x_axis], color='white', linestyle='dotted', label='Best Avg', alpha=0.8)

    print(f'Best {col_std} Performance: ')
    std_best = df.loc[df.loc[df[x_axis] != 1, col_std].idxmin()]
    print(std_best)
    plt.axvline(std_best[x_axis], color='green', linestyle='dotted', label='Best Std', alpha=0.8)

```
