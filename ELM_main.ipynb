```python
# Only for Classifier and int target
def cross_val_score(estimator, X, y=None, scoring=None, cv=None, stratified=True):
    if X is None: raise ValueError('X cannot be None')
    if y is None: raise ValueError('y cannot be None for supervised algorithm')
    if scoring is None: scoring = accuracy_score
    if cv is None: cv = min(5,pd.Series(y).value_counts().min()) # select the minimum possible folds or default of 5
    if pd.Series(y).value_counts().min() == 1 or cv == 1:
        raise ValueError('Cross Validation folds, cv, cannot be 1.')
    if stratified:
        kfold = StratifiedKFold(n_splits=cv, shuffle=False)
    else:
        kfold = KFold(n_splits=cv, shuffle=False)

    scores = []
    # get indices of each fold, and run model on each
    for i, (train_index, test_index) in enumerate(kfold.split(X, y)):
        # Fold i:
        x_train = X[train_index]
        y_train = y[train_index]
        x_test = X[test_index]
        y_test = y[test_index]
        estimator.fit(x_train, y_train)
        y_pred = estimator.predict(x_test)
        # print(confusion_matrix(y_test, y_pred))
        scores.append(scoring(y_test, y_pred))
    return scores
```

```python
from sklearn.metrics import log_loss, fbeta_score

neurons_cnt = range(1,201)
avrgs = []
maxs = []
mins = []
for i in neurons_cnt:
    ls = np.array(cross_val_score(ELMclf(i, random_state=42), X_train.to_numpy(), Y_train.to_numpy(), cv=5, scoring=(lambda x, y: f1_score(x, y, zero_division=0, average='binary'))))
    # ls = np.array(cross_val_score(ELMclf(i), X_train.to_numpy(), Y_train.to_numpy(), cv=5, scoring=(lambda x, y: roc_auc_score(x, y, average='macro'))))

    avrgs.append(ls.mean())
    maxs.append(ls.max())
    mins.append(ls.min())
```

```python
hidden_neuron_df = pd.DataFrame(np.transpose(np.array(
    [neurons_cnt,
     np.array(avrgs),
     np.array(maxs),
     np.array(mins)]
)), columns=['NeuronsCnt', 'Avg', 'Max', 'Min'])
axs = hidden_neuron_df.plot.scatter('NeuronsCnt', 'Avg', color='white')
hidden_neuron_df.plot.scatter('NeuronsCnt', 'Max', color='cyan', ax=axs)
hidden_neuron_df.plot.scatter('NeuronsCnt', 'Min', color='orange', ax=axs)
axs.plot(hidden_neuron_df['NeuronsCnt'], hidden_neuron_df['Max']-hidden_neuron_df['Min'], color='green')


print('Best Max performance: ')
max_best = hidden_neuron_df.loc[hidden_neuron_df['Max'].idxmax()]
print(max_best)
plt.axvline(max_best['NeuronsCnt'], color='cyan', linestyle='dotted', label='Best Max', alpha=0.8)

print('Best Min performance: ')
min_best = hidden_neuron_df.loc[hidden_neuron_df['Min'].idxmax()]
print(min_best)
plt.axvline(min_best['NeuronsCnt'], color='orange', linestyle='dashed', label='Best Min', alpha=0.5)

print('Best Avrg Performance: ')
avg_best = hidden_neuron_df.loc[hidden_neuron_df['Avg'].idxmax()]
print(avg_best)
plt.axvline(avg_best['NeuronsCnt'], color='white', linestyle='dotted', label='Best Avg', alpha=0.8)
```
