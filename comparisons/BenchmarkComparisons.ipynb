```python
def stat_is_success(df_res, thres=0.05):
    return df_res < thres

def show_test(df_x, df_y, test_func, params=None, use_stat=False):
    if params:
        if df_y is None:
            res = test_func(df_x, **params)
        else:
            res = test_func(df_x, df_y, **params)
    else:
        if df_y is None:
            res = test_func(df_x)
        else:
            res = test_func(df_x, df_y)
    if use_stat:
        res = res.statistic
        col = 'stat'
    else:
        res = res.pvalue
        col = 'pvalue'
    return pd.DataFrame(res, columns=[col], index=df_x.columns)

def show_wilcoxon(df_x, df_y, alternative='two-sided'):
    return show_test(df_x, df_y, wilcoxon, {'alternative': alternative})

def show_levene(df_x, df_y):
    # by default uses median as centre
    return show_test(df_x, df_y, levene)
```

```python
def performance_plot(df, colname_axis, ax):
    ax = sns.boxplot(
        df, color='cornflowerblue', ax=ax,
        medianprops=dict(color='red', linewidth=1.5))

    ax.set_title(f'{colname_axis} Performance of Models')
    ax.set_xlabel('Models', fontweight='bold')
    ax.set_ylabel(colname_axis, fontweight='bold')
    # ax.set_yticks(np.arange(0.87,0.97,0.01))
    ax.yaxis.set_minor_locator(MultipleLocator(0.005))
    ax.yaxis.set_major_locator(MultipleLocator(0.01))
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.grid(axis='y', which='major', linestyle='-', linewidth=0.75)
    ax.grid(axis='y', which='minor', linestyle=':', linewidth=0.5)
    return ax
```

```python
def stat_to_all(df, main_model_name, stat_func, params=None, use_stat=False):
    res_ls = []
    colname = 'stat' if use_stat else 'pvalue'
    for name in model_names:
        if name == main_model_name: continue
        res_ls.append(show_test(
            df.loc[main_model_name],
            df.loc[name],
            stat_func,
            params=params,
            use_stat=use_stat
        ).rename(columns={colname: name}))

    return pd.concat(res_ls, axis=1)

def stat_to_all_single(df, stat_func, params=None, use_stat=False):
    res_ls = []
    colname = 'stat' if use_stat else 'pvalue'
    for name in model_names:
        if params is not None:
            res_ls.append(show_test(df.loc[name], None, stat_func, params, use_stat).rename(columns={colname: name}))
        else:
            res_ls.append(show_test(df.loc[name], None, stat_func, use_stat=use_stat).rename(columns={colname: name}))
    return pd.concat(res_ls, axis=1)
```
