```python
import numpy as np
from matplotlib import pyplot as plt
from matplotlib.ticker import MultipleLocator
from scipy.stats import wilcoxon, levene, shapiro, ttest_ind
import pandas as pd

from utils import format_metric_df, subplot_txt
import seaborn as sns
```

```python
solu_cnt  = 15
max_iter  = 300
trial_lim = 30
```

```python
def stat_is_success(df_res, thres=0.05):
    return df_res < thres
```

```python
def show_test(df_x, df_y, test_func, params=None):
    if params: res = test_func(df_x, df_y, **params)
    else: res = test_func(df_x, df_y)
    return pd.DataFrame(res.pvalue, columns=['pvalue'], index=df_x.columns)
```

```python
def show_wilcoxon(df_x, df_y, alternative='two-sided'):
    return show_test(df_x, df_y, wilcoxon, {'alternative': alternative})
```

```python
def show_levene(df_x, df_y):
    # by default uses median as centre
    return show_test(df_x, df_y, levene)

# but this was omitted so it is no longer included and is useless
```

```python
# import the results for comparison
abc_relm_res = pd.read_csv(f'../important_data/abc_relm_results/result_single_abc_testres_sol{solu_cnt}_iter{max_iter}_lim{trial_lim}.csv', index_col=0)
abc_elm_res = pd.read_csv(f'../important_data/abc_elm_results/result_elm_single_abc_testres_sol{solu_cnt}_iter{max_iter}_lim{trial_lim}.csv', index_col=0)

elm_res  = pd.read_csv('../important_data/elm_results/result_elm_n38.csv', index_col=0)
relm_res = pd.read_csv('../important_data/relm_results/result_relm_n38_l36.csv', index_col=0)
```

```python
def performance_plot(df, colname_axis, ax):
    ax = sns.boxplot(
        df, color='cornflowerblue', ax=ax,
        medianprops=dict(color='red', linewidth=1.5))

    ax.set_title(f'{colname_axis} Performance of Models')
    ax.set_xlabel('Models', fontweight='bold')
    ax.set_ylabel(colname_axis, fontweight='bold')
    # ax.set_yticks(np.arange(0.87,0.97,0.01))
    ax.yaxis.set_minor_locator(MultipleLocator(0.005))
    ax.yaxis.set_major_locator(MultipleLocator(0.01))
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.grid(axis='y', which='major', linestyle='-', linewidth=0.75)
    ax.grid(axis='y', which='minor', linestyle=':', linewidth=0.5)
    return ax
```

```python
# Show boxplot
tick_size = None
axis_size = None
title_size = None

# this for multiple metric
fig, axs = plt.subplots(2,2,figsize=(12,8), layout='constrained')
colnames = ['precision', 'npv', 'recall', 'specificity']
colname_axiss = ['PPV', 'NPV', 'Sensitivity', 'Specificity']

tick_size = 12
axis_size = 14
title_size = 17

# chr 97 = 'a'
start = 97
plots_name = [f'({chr(i)})' for i in range(start, start+4)]

for i, (col, axname) in enumerate(zip(colnames, colname_axiss)):
    for_plot_elm = elm_res.copy().reset_index()
    for_plot_relm = relm_res.copy().reset_index()
    for_plot_abcelm = abc_elm_res.copy().reset_index()
    for_plot_abcrelm = abc_relm_res.copy().reset_index()
    for_plot_elm.rename(columns={col:'ELM'}, inplace=True)
    for_plot_relm.rename(columns={col: 'RELM'}, inplace=True)
    for_plot_abcelm.rename(columns={col: 'ABC-ELM'}, inplace=True)
    for_plot_abcrelm.rename(columns={col: 'ABC-RELM'}, inplace=True)
    combined = pd.concat([for_plot_elm['ELM'],for_plot_relm['RELM'],for_plot_abcelm['ABC-ELM'], for_plot_abcrelm['ABC-RELM']], axis=1)

    is_subplot = isinstance(axs, np.ndarray)

    ax = axs.ravel()[i] if is_subplot else axs
    boxplot = performance_plot(combined, axname, ax=ax)
    boxplot.set_title(ax.get_title(), size=title_size)
    boxplot.xaxis.set_tick_params(labelsize=tick_size)
    boxplot.yaxis.set_tick_params(labelsize=tick_size)
    boxplot.set_xlabel(boxplot.get_xlabel(), fontsize=axis_size)
    boxplot.set_ylabel(boxplot.get_ylabel(), fontsize=axis_size)
    if is_subplot:
        txt = subplot_txt(plots_name[i], ax)

        # txt = ax.text(-0,1.05, plots_name[i], ha='left', va='top', transform=ax.transAxes)
        txt.set_in_layout(False)

# plt.tight_layout()
plt.show()
```

# Cross Validation Section

```python
def fold_comparison(df_x, df_y, fold=0, alternative=None):
    if fold == 0:
        return show_wilcoxon(df_x, df_y, alternative=alternative)
    return show_wilcoxon(df_x.loc[df_x['Fold'] == fold].drop('Fold', axis=1), df_y.loc[df_y['Fold'] == fold].drop('Fold', axis=1), alternative=alternative)
```

```python
cv_abc_elm_res = pd.read_csv(f'../important_data/abc_elm_results/result_elm_abc_scores_sol{solu_cnt}_iter{max_iter}_lim{trial_lim}.csv', index_col=0)
cv_abc_relm_res = pd.read_csv(f'../important_data/abc_relm_results/result_abc_scores_sol{solu_cnt}_iter{max_iter}_lim{trial_lim}.csv', index_col=0)

cv_elm_res = pd.read_csv(f'../important_data/elm_results/result_cv_elm_n38.csv', index_col=0).drop('Seed', axis=1)
cv_relm_res = pd.read_csv(f'../important_data/relm_results/result_cv_relm_n38_l36.csv', index_col=0).drop('Seed', axis=1)

cv_abc_elm_res_train = cv_abc_elm_res.loc[cv_abc_elm_res['predicted_data'] == 'train_set'].drop('predicted_data', axis=1).drop('Seed', axis=1)
cv_abc_elm_res_test  = cv_abc_elm_res.loc[cv_abc_elm_res['predicted_data'] == 'test_set'].drop('predicted_data', axis=1).drop('Seed', axis=1)
cv_abc_relm_res_train = cv_abc_relm_res.loc[cv_abc_relm_res['predicted_data'] == 'train_set'].drop('predicted_data', axis=1).drop('Seed', axis=1)
cv_abc_relm_res_test  = cv_abc_relm_res.loc[cv_abc_relm_res['predicted_data'] == 'test_set'].drop('predicted_data', axis=1).drop('Seed', axis=1)
```

```python
stat_is_success(show_wilcoxon(cv_abc_relm_res_test.loc[cv_abc_relm_res_test['Fold'] == 1].drop('Fold', axis=1), cv_elm_res.loc[cv_elm_res['Fold'] == 1].drop('Fold', axis=1), 'greater'))
```

```python
fold_comparison(cv_abc_relm_res_test, cv_elm_res, 1, 'greater')
```

```python
fold_pvalues = [fold_comparison(cv_abc_relm_res_test, cv_relm_res, fold, 'greater') for fold in range(1,6)]
fold_pvalues_df = pd.concat(fold_pvalues, axis=1)
```
