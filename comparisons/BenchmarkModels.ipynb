```python
import numpy as np
import pandas as pd
from scipy.stats import uniform
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier

from preprocess import get_data, preprocess
from utils import get_metrics, cross_val_score
```

Use Cross Validation Grid Search to find optimal hyperparameters for the benchmark models, then run them across the same random seed range and store the results into csv.

```python
split_seed = 42

# Importing preprocessed dataset
df = get_data('../data/thyroid_diff.csv')
df = preprocess(df, do_onehot=True, do_scale=True)

X, Y = df.drop('Recurred_Yes', axis=1), df['Recurred_Yes']
X_np, Y_np = np.asarray(X, dtype=np.float32), np.asarray(Y, dtype=np.float32)
x_train,     x_test,     y_train,     y_test     = train_test_split(X, Y, test_size=0.2, random_state=split_seed, stratify=Y)
# x_train, x_test, y_train, y_test = (
#     x_train.astype(np.float32),
#     x_test.astype(np.float32),
#     y_train.astype(np.float32),
#     y_test.astype(np.float32))
x_train_sub, x_validate, y_train_sub, y_validate = train_test_split(x_train, y_train, test_size=0.25, random_state=split_seed, stratify=y_train)
```

```python
r_seed = 42
# linear
lr  = LogisticRegression(solver='liblinear') # l1 log reg

# can use kernel
svc = SVC()

# tree
dt  = DecisionTreeClassifier(criterion='entropy') # for balances given by it

# data based
knn = KNeighborsClassifier(algorithm='kd_tree', p=2)

# ensemble
rf  = RandomForestClassifier(criterion='entropy')
gb  = GradientBoostingClassifier(loss='log_loss')

# neural network
mlp = MLPClassifier(solver='adam', activation='logistic')

models = [lr, svc, dt, knn, rf, gb, mlp]
for model in models:
    model.random_state = r_seed
```

```python
lr_dist = dict(
    C=uniform(loc=10e-4, scale=10e4),
    max_iter=range(100,1001),
)
svc_dist = dict(
    C=uniform(loc=10e-4, scale=10e4),
    kernel=['linear', 'rbf', 'poly', 'sigmoid'],
    degree=[2, 3, 4],
    max_iter=range(100,1001),
)
dt_dist = dict(
    max_depth=range(1,30),
    min_samples_split=range(2,10),
    min_samples_leaf=range(1,10),
    max_features=range(1,x_train.shape[1]+1),
)
knn_dist = dict(
    n_neighbors=range(3,21,2),
    leaf_size=range(1,50),
)
rf_dist = dict(
    n_estimators=range(100,501,10),
    max_depth=range(1,30),
    min_samples_split=range(2,10),
    min_samples_leaf=range(1,10),
    max_features=range(1,x_train.shape[1]+1),
)
gb_dist = dict(
    n_estimators=range(100,501,10),
    learning_rate=uniform(loc=0, scale=5),
    subsample=uniform(loc=0, scale=1),
    min_samples_split=range(2,10),
    min_samples_leaf=range(1,10),
    max_depth=range(1,30),
)
mlp_dist = dict(
    hidden_layer_sizes=[(i,) for i in range(1,80)],
    alpha=uniform(loc=0, scale=1),
    max_iter=range(100,1001),
)


param_dists = {
    f'{lr.__class__.__name__}': lr_dist,
    f'{svc.__class__.__name__}': svc_dist,
    f'{dt.__class__.__name__}': dt_dist,
    f'{knn.__class__.__name__}': knn_dist,
    f'{rf.__class__.__name__}': rf_dist,
    f'{gb.__class__.__name__}': gb_dist,
    f'{mlp.__class__.__name__}': mlp_dist,
}
best_params = {}
for model in models:
    name = model.__class__.__name__
    dist = param_dists.get(name, None)
    if dist is None:
        raise Exception(f'Model distributions not found in param_dist, or the value is None: {name}')
    rs = RandomizedSearchCV(model, dist, cv=5, random_state=r_seed, n_jobs=7, n_iter=10)
    rs = rs.fit(x_train, y_train)
    best_params[name] = rs.best_params_
```

```python
for model in models:
    print(model.__class__.__name__, best_params[model.__class__.__name__],'\n')
```

```python
def one_run_of_all(r_state):
    res_list = []
    for model in models:
        name = model.__class__.__name__
        params = best_params.get(name, None)
        if params is None:
            raise Exception(f'Model params not found in best_params, or the value is None: {name}')
        model = model.__class__(**params)
        model.random_state = r_state
        model.fit(x_train, y_train)

        res = get_metrics(y_test, model.predict(x_test))
        res = {'model': name} | {'seed': r_state} | res
        res_list.append(res)
    return pd.DataFrame(res_list)

big_res = [one_run_of_all(r_state) for r_state in range(0, 50)]
```

```python
indi_res = {}
big_test_df = pd.concat(big_res)
for model in models:
    name = model.__class__.__name__
    indi_res[name] = big_test_df.loc[big_test_df['model'] == name].drop('model', axis=1)
```

```python
def one_cv_of_all(r_state):
    res_list = []
    for model in models:
        name = model.__class__.__name__
        params = best_params.get(name, None)
        if params is None:
            raise Exception(f'Model params not found in best_params, or the value is None: {name}')
        model = model.__class__(**params)
        model.random_state = r_state
        res_folds = cross_val_score(model, X.to_numpy(), Y.to_numpy(), cv=5, scoring=get_metrics)
        # model.fit(x_train, y_train)
        # res = get_metrics(y_test, model.predict(x_test))
        res = pd.DataFrame(res_folds)
        res.insert(0, 'Fold', [1,2,3,4,5])
        res.insert(0, 'Seed', r_state)
        res.insert(0, 'Model', name)
        res_list.append(res)
    return pd.concat(res_list)

big_cv_res = [one_cv_of_all(r_state) for r_state in range(0, 50)]
```

```python
indi_cv_res = {}
big_cv_df = pd.concat(big_cv_res)
for model in models:
    name = model.__class__.__name__
    indi_cv_res[name] = big_cv_df.loc[big_cv_df['Model'] == name].drop('Model', axis=1)
```

```python
for key, value in indi_cv_res.items():
    value.to_csv(f'../data/benchmark_cv_{key}.csv')
```
