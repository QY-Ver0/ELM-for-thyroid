```python
import copy

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, fbeta_score, confusion_matrix
import numpy as np
from ELM import ELMclf
import pandas as pd
from sklearn.model_selection import train_test_split, StratifiedKFold, KFold
from sklearn.preprocessing import MinMaxScaler
from preprocess import preprocess, get_XY, get_data
from optimiser import Optimiser, ABCOptimiser, OptimiserName, cross_val_optimisation
from utils import get_metrics_df, cross_val_score, get_metrics, train_test_graph
```

```python
def multi_seed_cross_val_abc(optimiser: Optimiser, X, Y, args, cv_scoring=accuracy_score, cv=5, split_seed=42, ret_graph_df=True, val_size=0.25, seed_range=range(0,102,2)):
    # val_size: validation size for split in train set of validation set
    ret_train = True
    optimiser_copy = copy.deepcopy(optimiser)

    final_df = None
    final_scores_df = None
    for i in seed_range:
        print(f'Seed {i}:')
        optimiser_copy.random_state = i
        train_scores, scores, dflist, scores_dflist = cross_val_score(optimiser_copy, X, Y, scoring=cv_scoring, cv=cv, args=args, random_state=split_seed, ret_train=ret_train, ret_graph_df=ret_graph_df, val_size=val_size)
        full_dflist = pd.concat(dflist)
        full_scores_dflist = pd.concat(scores_dflist)

        print(full_dflist.shape, len(dflist))

        full_dflist.insert(0, 'Seed', np.repeat(i, full_dflist.shape[0]))
        full_scores_dflist.insert(0, 'Seed', np.repeat(i, full_scores_dflist.shape[0]))
        if final_df is None and final_scores_df is None:
            final_df = full_dflist
            final_scores_df = full_scores_dflist
        else:
            final_df = pd.concat([final_df, full_dflist]).reset_index(drop=True)
            final_scores_df = pd.concat([final_scores_df, full_scores_dflist]).reset_index(drop=True)
    return final_df, final_scores_df
```

```python
# setting params
n            = 31
# l2           = 0.5625
l2           = 0.4
def_rng_seed = 42
split_seed   = 42
global_rng   = np.random.default_rng(def_rng_seed)

solution_cnt = 15
# defined_fn   = lambda yt, yp: f1_score(yt, yp, average='micro')
defined_fn   = f1_score
max_iter     = 200 # 10 solution, 1000 max iter
trial_limit  = 20
cv           = 5

# 5 - 200 - 40 : abc_elm_v2100
# 10 - 1000 - 40 : abc_elm100
# 10 - 150 - 15 : Single abc_elm best outcome
```

```python
# Importing preprocessed dataset
# df = pd.read_csv('data_preprocessed.csv', index_col=0)
df = get_data('data/thyroid_diff.csv')
df = preprocess(df, do_onehot=True, do_scale=True)

X, Y = df.drop('Recurred_Yes', axis=1), df['Recurred_Yes']
X_np, Y_np = X.to_numpy(), Y.to_numpy()
x_train,     x_test,     y_train,     y_test     = train_test_split(X, Y, test_size=0.2, random_state=split_seed, stratify=Y)
x_train_sub, x_validate, y_train_sub, y_validate = train_test_split(x_train, y_train, test_size=0.25, random_state=split_seed, stratify=y_train)

# x_train_sc, x_test_sc, y_train_sc, y_test_sc = x_train, x_test, y_train, y_test
# x_train_sub_sc, x_validate_sc, y_train_sub_sc, y_validate_sc = x_train_sub, x_validate, y_train_sub, y_validate
```

```python
# Getting initial solutions
elm = ELMclf(n, l2_param=l2, random_state=def_rng_seed)
elm.fit(x_train_sub.to_numpy(), y_train_sub.to_numpy())

popu_temp  = {'weights_i': elm.weights_i, 'biases': elm.biases} # template
popu_range = {'weights_i': (-1,1),        'biases': (-1,1)}     # range
popu = Optimiser.population_generation(solution_cnt, popu_temp, popu_range, random_state=int(global_rng.uniform(0, 4294967295)))
abc_args = {
    'population':       popu,
    'population_limit': popu_range,
    'max_iter':         max_iter,
    'trial_limit':      trial_limit,
    'cv':               cv
}
abc_args = {
    'population':       popu,
    'population_limit': popu_range,
    'max_iter':         20,
    'trial_limit':      trial_limit,
    'cv':               cv
}
```

```python
# Init abc optimiser
# abc = ABCOptimiser(elm, x_train_sub.to_numpy(), y_train_sub.to_numpy(), x_validate.to_numpy(), y_validate.to_numpy(), fitness_fn=defined_fn, random_state=def_rng_seed)
abc = ABCOptimiser(elm, fitness_fn=defined_fn, random_state=def_rng_seed)
abc.set_verbose(True)
```
