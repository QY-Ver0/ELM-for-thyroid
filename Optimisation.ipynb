```python
import json

import numpy as np
from ELM import ELMclf
import pandas as pd
from sklearn.model_selection import train_test_split

from preprocess import preprocess, get_data
from optimiser_main import Optimiser, ABCOptimiser, as_solution, as_params, get_dims
from utils import get_metrics_df, cross_val_score, get_metrics, train_test_graph, f1
```

```python
# setting params
n            = 38
l2           = 0.36
def_rng_seed = 42
split_seed   = 42
global_rng   = np.random.default_rng(def_rng_seed)
defined_fn   = f1
cv           = 5

solution_cnt = 15
max_iter     = 300
trial_limit  = solution_cnt * 2
```

```python
# Importing preprocessed dataset
df = get_data('data/thyroid_diff.csv')
df = preprocess(df, do_onehot=True, do_scale=True)

X, Y = df.drop('Recurred_Yes', axis=1), df['Recurred_Yes']
X_np, Y_np = np.asarray(X, dtype=np.float32), np.asarray(Y, dtype=np.float32)
x_train,     x_test,     y_train,     y_test     = train_test_split(X, Y, test_size=0.2, random_state=split_seed, stratify=Y)
x_train_sub, x_validate, y_train_sub, y_validate = train_test_split(x_train, y_train, test_size=0.125, random_state=split_seed, stratify=y_train)
```

```python
# Getting initial solutions
elm = ELMclf(n, l2_param=l2, random_state=def_rng_seed)
elm.fit(x_train_sub.to_numpy(), y_train_sub.to_numpy())

popu_temp  = elm.return_args() # template
popu_temp.pop('l2_param')
dims = get_dims(popu_temp)
popu_range = {'weights_i': (-1,1), 'biases': (-1,1)}     # range
popu = Optimiser.population_generation(solution_cnt, popu_temp, popu_range, random_state=int(global_rng.uniform(0, 4294967295)))
abc_args = {
    'population':       popu,
    'population_limit': popu_range,
    'max_iter':         max_iter,
    'trial_limit':      trial_limit,
    'cv':               5,
    'cv_on_train':      False,
}
```

```python
# Init abc optimiser
abc = ABCOptimiser(elm, fitness_fn=defined_fn, random_state=def_rng_seed)
abc.set_verbose(True)
```

```python
best_indi = abc.fit(x_train.to_numpy(), y_train.to_numpy(), x_test.to_numpy(), y_test.to_numpy(), **abc_args) # cont
```

```python
train_res_i = abc.train_res
# train_res_i = abc.vali_res
val_res_i   = abc.vali_res
# val_res_i   = abc.test_res
scouts_cnt  = abc.lim_reached_cnts

# append DataFrame list
tt_df = pd.DataFrame([train_res_i, val_res_i, scouts_cnt], index=['Train', 'Validation', 'ScoutCall'])
tt_df = tt_df.transpose()
iters = range(1, tt_df.shape[0] + 1)
tt_df.insert(0, 'Iters', iters)
# tt_df.to_csv(f'data/final_abc_sol{solution_cnt}_iter{max_iter}_lim{trial_limit}.csv')
tt_df.to_csv(f'data/final_t_abc_sol{solution_cnt}_iter{max_iter}_lim{trial_limit}.csv')
```

```python
scores_val_df = get_metrics_df(
    abc.models[abc.best_idx].fit(x_train.to_numpy(), y_train.to_numpy(), **best_indi),
    x_train_sub.to_numpy(),
    x_validate.to_numpy(),
    y_train_sub.to_numpy(), y_validate.to_numpy())
# scores_val_df.to_csv(f'data/final_abc_scores_sol{solution_cnt}_iter{max_iter}_lim{trial_limit}.csv')
scores_val_df.to_csv(f'data/final_t_abc_scores_sol{solution_cnt}_iter{max_iter}_lim{trial_limit}.csv')
```
