```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, fbeta_score, confusion_matrix
import numpy as np
from ELM import ELMclf
import pandas as pd
from sklearn.model_selection import train_test_split, StratifiedKFold, KFold
from optimiser import Optimiser, ABCOptimiser
```

```python
# Some functions used
def cross_val_score(estimator, X, y=None, scoring=None, cv=None, shuffle=False, stratified=True, args=None):
    if X is None: raise ValueError('X cannot be None')
    if y is None: raise ValueError('y cannot be None for supervised algorithm')
    if scoring is None: scoring = accuracy_score
    if cv is None: cv = min(5,pd.Series(y).value_counts().min()) # select the minimum possible folds or default of 5
    if pd.Series(y).value_counts().min() == 1 or cv == 1:
        raise ValueError('Cross Validation folds, cv, cannot be 1.')
    if stratified:
        kfold = StratifiedKFold(n_splits=cv, shuffle=shuffle)
    else:
        kfold = KFold(n_splits=cv, shuffle=shuffle)

    scores = []
    # get indices of each fold, and run model on each
    for i, (train_index, test_index) in enumerate(kfold.split(X, y)):
        # Fold i:
        x_train = X[train_index]
        y_train = y[train_index]
        x_test  = X[test_index]
        y_test  = y[test_index]
        if args:
            estimator.fit(x_train, y_train, **args)
        else:
            estimator.fit(x_train, y_train)
        y_pred = estimator.predict(x_test)
        # print(confusion_matrix(y_test, y_pred))
        scores.append(scoring(y_test, y_pred))
    return scores


def get_metrics(y, y_pred):
    # to get pandas dataframe for showing a table of comparison
    # between train result and test result
    return {
        'accuracy': accuracy_score(y, y_pred),
        'precision': precision_score(y, y_pred, average='binary', zero_division=0),
        'recall': recall_score(y, y_pred, average='binary', zero_division=0),
        # 'specificity': specificity_score(y, y_pred, zero_division=0),
        'f1': f1_score(y, y_pred, average='binary', zero_division=0),
        'f2': fbeta_score(y, y_pred, beta=2, average='binary', zero_division=0),
    }

def get_metrics_df(model_fitted: ELMclf, x_train, x_test, y_train, y_test):
    # Assumes model to be fitted
    train_pred = model_fitted.predict(x_train.to_numpy())
    test_pred = model_fitted.predict(x_test.to_numpy())
    res = []

    for true, pred in [(y_train, train_pred), (y_test, test_pred)]:
        res.append(get_metrics(true,pred))

    metrics_df = pd.DataFrame(res)
    metrics_df = pd.concat((pd.DataFrame(['train_set','test_set'], columns=['predicted_data']), metrics_df), axis=1)

    return metrics_df
```

```python
# setting params
n            = 58
l2           = 0.015625
def_rng_seed = 42
split_seed   = 42
global_rng   = np.random.default_rng(def_rng_seed)

solution_cnt = 2
defined_fn   = f1_score
max_iter     = 1000
trial_limit  = 40
```

```python
# Importing preprocessed dataset
df = pd.read_csv('data_preprocessed.csv', index_col=0)
booleans = df.drop('Age', axis=1)
df[booleans.columns] = booleans.astype('int64')
```

```python
# Obtaining data splits
X = df.drop('Recurred_Yes', axis=1)
Y = df['Recurred_Yes']
x_train,     x_test,     y_train,     y_test     = train_test_split(X, Y, test_size=0.2, random_state=split_seed, stratify=Y)
x_train_sub, x_validate, y_train_sub, y_validate = train_test_split(x_train, y_train, test_size=0.25, random_state=split_seed, stratify=y_train)
```

```python
# Getting initial solutions
elm = ELMclf(n, l2_param=l2, random_state=def_rng_seed)
elm.fit(x_train_sub.to_numpy(), y_train_sub.to_numpy())

popu_temp  = {'weights_i': elm.weights_i, 'biases': elm.biases} # template
popu_range = {'weights_i': (-1,1),        'biases': (-1,1)}     # range
popu = Optimiser.population_generation(solution_cnt, popu_temp, popu_range, random_state=int(global_rng.uniform(0, 4294967295)))
```

```python
# Init abc optimiser
abc = ABCOptimiser(elm, x_train_sub.to_numpy(), y_train_sub.to_numpy(), x_validate.to_numpy(), y_validate.to_numpy(), fitness_fn=defined_fn, random_state=def_rng_seed)
abc.set_verbose(True)
```

```python
print('Original Validation Results:')
get_metrics_df(elm.fit(x_train_sub.to_numpy(), y_train_sub.to_numpy()), x_train_sub, x_validate, y_train_sub, y_validate)
```

```python
print('Original Test Results:')
# elm = ELMclf(n, l2_param=l2, random_state=def_rng_seed)
get_metrics_df(elm.fit(x_train.to_numpy(), y_train.to_numpy()), x_train, x_test, y_train, y_test)
```

```python
# Running optimiser
best_param = abc.fit(popu, popu_range, max_iter=max_iter, trial_limit=trial_limit, cv=5)
abc.score()
```

```python
# ABC Results
abc_elm = ELMclf(n, l2_param=l2, random_state=def_rng_seed)
print('ABC-ELM Results: ')
get_metrics_df(abc_elm.fit(x_train.to_numpy(), y_train.to_numpy(), **best_param), x_train, x_test, y_train, y_test)
```

```python
# ABC Cross Validation Results
scores = cross_val_score(abc_elm, X.to_numpy(), Y.to_numpy(), scoring=get_metrics, cv=5)
pd.DataFrame(scores).mean()
```

```python
# Random seed check for ABC algo
res = []
abc_elm_check = ELMclf(n, l2_param=l2, random_state=def_rng_seed) # this is useless here
for r in range(0,50):
    print(f'ABC with random seed {r}')
    popu = Optimiser.population_generation(solution_cnt, popu_temp, popu_range, random_state=r)
    abc = ABCOptimiser(elm, x_train_sub.to_numpy(), y_train_sub.to_numpy(), x_validate.to_numpy(), y_validate.to_numpy(), fitness_fn=defined_fn, random_state=r)
    abc.set_verbose(True)
    best_param = abc.fit(popu, popu_range, max_iter=max_iter, trial_limit=trial_limit, cv=5)

    abc_elm_check.fit(x_train.to_numpy(), y_train.to_numpy(), **best_param)
    res.append(get_metrics(y_test.to_numpy(), abc_elm_check.predict(x_test.to_numpy())))

res_df = pd.DataFrame(res)
res
```
