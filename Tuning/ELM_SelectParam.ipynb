```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from matplotlib.ticker import MultipleLocator
# scikit-learn == 1.5.2
from sklearn.model_selection import train_test_split

from ELM import ELMclf
from preprocess import preprocess, get_data

from utils import cross_val_score, get_metrics, format_metric_df, subplot_txt
```

```python
split_seed = 42
```

```python
path_to_data = '../data/'
path_to_data_imp = '../important_data/'
```

```python
# Acquire preprocessed dataset
df = get_data(f'{path_to_data}thyroid_diff.csv')
df = preprocess(df, do_onehot=True, do_scale=True)

X = df.drop('Recurred_Yes', axis=1)
Y = df['Recurred_Yes']
x_train,     x_test,     y_train,     y_test     = train_test_split(X, Y, test_size=0.2, random_state=split_seed, stratify=Y)
x_train_sub, x_validate, y_train_sub, y_validate = train_test_split(x_train, y_train, test_size=0.25, random_state=split_seed, stratify=y_train)
```

```python
def plot_with(df_param, title=None, use_scatter=True, ylabel='F1', showwhichcol=None, xaxis_scale=None, yaxis_scale=None, y_lim=None, show_std=True, isint=False, select_highest=True, color='cornflowerblue', ax=None, bbox=(0.5, -0.24)):

    x_axis, col_avg, col_max, col_min, col_std = df_param.columns
    if use_scatter:
        axs = df_param.plot.scatter(x_axis, col_avg, color=color, title=title, ax=ax)
        df_param.plot.scatter(x_axis, col_max, color='cyan', ax=axs)
        df_param.plot.scatter(x_axis, col_min, color='orange', ax=axs)
    else:
        axs = df_param.plot.line(x_axis, col_avg, color=color, title=title, ax=ax)
        # df.plot.line(x_axis, col_max, color='cyan', ax=axs, alpha=0.4)
        # df.plot.line(x_axis, col_min, color='orange', ax=axs, alpha=0.4)
    # df.plot.line(x_axis, col_std, color='green', ax=axs)
    axs.margins(x=0.01,y=0.01)
    if xaxis_scale is not None:
        if xaxis_scale[0] is not None: axs.xaxis.set_major_locator(MultipleLocator(xaxis_scale[0]))
        if xaxis_scale[1] is not None: axs.xaxis.set_minor_locator(MultipleLocator(xaxis_scale[1]))
    if yaxis_scale is not None:
        if yaxis_scale[0] is not None: axs.yaxis.set_major_locator(MultipleLocator(yaxis_scale[0]))
        if yaxis_scale[1] is not None: axs.yaxis.set_minor_locator(MultipleLocator(yaxis_scale[1]))

    if y_lim is not None:
        axs.set_ylim(*y_lim)
    axs.grid(which='major', linestyle=':', linewidth=0.5)
    axs.spines['top'].set_visible(False)
    axs.spines['right'].set_visible(False)
    axs.set_xlabel(axs.get_xlabel(), fontweight='bold')
    axs.set_ylabel(ylabel, fontweight='bold')

    if show_std:
        plt.fill_between(df_param[x_axis], df_param[col_avg] - df_param[col_std], df_param[col_avg] + df_param[col_std], alpha=0.25, label='Std') # std area
    # df.plot.line('NeuronsCnt', 'Avg/Std', color='green')
    # axs.plot(df['NeuronsCnt'], df['Max']-df['Min'], color='green')


    print(f'Best {col_max} performance: ')
    max_best = df_param.loc[df_param[col_max].idxmax()]
    print(max_best)
    if showwhichcol == 2:
        plt.axvline(max_best[x_axis], color=color, linestyle='dotted', label=f'Best {x_axis}: {int(max_best[x_axis]) if isint else max_best[x_axis]}', alpha=0.8)
        # plt.axvline(max_best[x_axis], color='cyan', linestyle='dotted', label='Best Max', alpha=0.8)

    print(f'Best {col_min} performance: ')
    min_best = df_param.loc[df_param[col_min].idxmax()]
    print(min_best)
    if showwhichcol == 3:
        plt.axvline(min_best[x_axis], color=color, linestyle='dotted', label=f'Best {x_axis}: {int(min_best[x_axis]) if isint else min_best[x_axis]}', alpha=0.8)
        # plt.axvline(min_best[x_axis], color='orange', linestyle='dashed', label='Best Min', alpha=0.5)

    print(f'Best {col_avg} Performance: ')
    if select_highest:
        avg_best = df_param.loc[df_param[col_avg].idxmax()]
    else:
        avg_best = df_param.loc[df_param[col_avg].idxmin()]
    print(avg_best)
    if showwhichcol == 1:
        plt.axvline(avg_best[x_axis], color=color, linestyle='dotted', label=f'Best {x_axis}: {int(avg_best[x_axis]) if isint else avg_best[x_axis]}', alpha=0.8)

    print(f'Best {col_std} Performance: ')
    std_best = df_param.loc[df_param.loc[df_param[x_axis] != 1, col_std].idxmin()]
    print(std_best)
    if showwhichcol == 4:
        plt.axvline(std_best[x_axis], color=color, linestyle='dotted', label=f'Best {x_axis}: {int(std_best[x_axis]) if isint else std_best[x_axis]}', alpha=0.8)
        # plt.axvline(std_best[x_axis], color='green', linestyle='dotted', label='Best Std', alpha=0.8)

    ncol = len(axs.get_legend_handles_labels()[1])
    axs.legend(loc='lower center', bbox_to_anchor=bbox, ncol=ncol, frameon=False, fontsize=10)
    return axs

```

```python
all_neuron_multimodel_df = pd.read_csv(f'{path_to_data}node2_100__seed0_100_2__l0_1_004.csv')
```

```python
l0_df = all_neuron_multimodel_df.loc[all_neuron_multimodel_df['L2'] == 0]
meaned_l0_df = l0_df.groupby('NeuronsCnt', as_index=False).mean()
elm_best_neuroncnt = meaned_l0_df.loc[meaned_l0_df['F1_Avg'].idxmax(), 'NeuronsCnt']
for_plot = meaned_l0_df.rename(columns={'NeuronsCnt': 'Hidden Node Size'})
plot_with(for_plot[['Hidden Node Size', 'F1_Avg', 'F1_Max', 'F1_Min', 'F1_Std']], 'F1 for Each Hidden Node Size', use_scatter=False, showwhichcol=1, xaxis_scale=(10,5), yaxis_scale=(None,0.05), isint=True)
```

```python
meaned_n_df = all_neuron_multimodel_df.loc[all_neuron_multimodel_df['NeuronsCnt'] == elm_best_neuroncnt].groupby('L2', as_index=False).mean()
for_plot = meaned_n_df.copy()
for_plot['F1_Avg_Lower'] = for_plot['F1_Avg'] - for_plot['F1_Std']

yscale = (0.001,0.0005)
# fig, (top, bottom, std_plot) = plt.subplots(3,1, gridspec_kw={'height_ratios': [4, 6, 3]})
fig, (top, bottom) = plt.subplots(2,1, gridspec_kw={'height_ratios': [2,3]})
plt.subplots_adjust(wspace=0.4, hspace=0.6)


plot_with(for_plot[['L2', 'F1_Avg_Lower', 'F1_Max', 'F1_Min', 'F1_Std']], use_scatter=False, showwhichcol=1, xaxis_scale=(None,0.04), yaxis_scale=yscale,  show_std=False, color='lightblue', ax=bottom, bbox=(0.5, -0.45))
plot_with(meaned_n_df[['L2', 'F1_Avg', 'F1_Max', 'F1_Min', 'F1_Std']], f'F1 at Each L2 Parameter', use_scatter=False, showwhichcol=0, xaxis_scale=(None,0.04), yaxis_scale=yscale, y_lim=(None, 0.913),show_std=False, ax=top, bbox=(0.5, -0.65))

# top.set_xlabel('')
subplot_txt('(a)', top, -0.12, 1.2, fontsize='small')
subplot_txt('(b)', bottom, -0.12, 1.15, fontsize='small')

changed_std_df = meaned_n_df.copy()
changed_std_df.rename(columns={'F1_Std': 'F1_SD'}, inplace=True)
std_plot = plot_with(changed_std_df[['L2', 'F1_SD', 'F1_Max', 'F1_Min', 'F1_Avg']], 'Std at Each L2 Parameter', use_scatter=False, ylabel='Std', showwhichcol=0, xaxis_scale=(None,0.04), yaxis_scale=(0.001,0.0005), y_lim=(0.025,0.029), show_std=False, select_highest=False, color='green', bbox=(0.5, -0.6))
std_plot.set_box_aspect(0.25)
# relm_best_l2 = meaned_n_df.loc[meaned_n_df['F1_Std'].idxmin(), 'L2']
relm_best_l2 = for_plot.loc[for_plot['F1_Avg_Lower'].idxmax(), 'L2']
# fig.tight_layout()
```

# <b>Run To Predict Test set Across Multiple Seeds</b>

```python
r_seed = 42
elm1 = ELMclf(int(elm_best_neuroncnt), l2_param=0, random_state=r_seed)
elm2 = ELMclf(int(elm_best_neuroncnt), l2_param=relm_best_l2, random_state=r_seed)

train_res  = []
res        = []
test_preds = []

for elm in [elm1, elm2]:
    elm_desc = {'NeuronsCnt': elm.n, 'L2': elm.l2_param}
    elm.fit(x_train.to_numpy(), y_train.to_numpy())

    y_pred_train = elm.predict(x_train.to_numpy())
    train_metrics = elm_desc | get_metrics(y_train.to_numpy(), np.array(y_pred_train))
    train_res.append(train_metrics)
    # print(confusion_matrix(y_train.to_numpy(), y_pred_train))


    y_pred = elm.predict(x_test.to_numpy())
    outer_metrics = elm_desc | get_metrics(y_test.to_numpy(), np.array(y_pred))
    res.append(outer_metrics)
    test_preds.append(y_pred)
    # print(confusion_matrix(y_test.to_numpy(), y_pred))

train_metrics_df = pd.DataFrame(train_res)
train_metrics_df = pd.concat(
    (pd.DataFrame([
        f' ELM with {int(elm_best_neuroncnt) } neurons',
        f'RELM with {int(elm_best_neuroncnt) } neurons, L2: {relm_best_l2}'
    ], columns=['Description']), train_metrics_df
    ), axis=1)


metrics_df = pd.DataFrame(res)
metrics_df = pd.concat(
    (pd.DataFrame([
        f' ELM with {int(elm_best_neuroncnt) } neurons',
        f'RELM with {int(elm_best_neuroncnt) } neurons, L2: {relm_best_l2}'
    ], columns=['Description']), metrics_df
    ), axis=1)

print('Results:')
for_plot = metrics_df.copy()
pd.set_option('display.float_format', '{:.4f}'.format)
format_metric_df(for_plot.drop(['NeuronsCnt', 'L2'],axis=1)).reindex(columns=['Description', 'F1', 'PPV', 'NPV', 'TPR', 'TNR'])
```

```python
# multi seed check
big_train_res = []
big_res       = []
big_train_multifold = []
big_test_multifold  = []

elm1 = ELMclf(int(elm_best_neuroncnt), l2_param=0, random_state=r_seed)
elm2 = ELMclf(int(elm_best_neuroncnt), l2_param=relm_best_l2, random_state=r_seed)

cv_cnt = 5

for i in range(0,50):
    train_res  = []
    res        = []
    test_preds = []

    # Run for each ELM model
    for elm_param in [
        {'n': elm1.n, 'l2_param': elm1.l2_param},
        {'n': elm2.n, 'l2_param': elm2.l2_param}]:

        elm      = ELMclf(random_state=i, **elm_param)
        elm_desc = {'NeuronsCnt': elm.n, 'L2': elm.l2_param}
        elm.fit(x_train.to_numpy(), y_train.to_numpy())

        # Predict Train
        y_pred_train  = elm.predict(x_train.to_numpy())
        train_metrics = elm_desc | get_metrics(y_train.to_numpy(), np.array(y_pred_train))
        train_res.append(train_metrics)

        # Predict Test
        y_pred        = elm.predict(x_test.to_numpy())
        outer_metrics = elm_desc | get_metrics(y_test.to_numpy(), np.array(y_pred))
        res.append(outer_metrics)
        test_preds.append(y_pred)

        # Cross Validation
        train_cv, test_cv = cross_val_score(elm, X.to_numpy(), Y.to_numpy(), scoring=get_metrics, cv=cv_cnt, ret_train=True)

        train_df = pd.DataFrame(train_cv)
        train_df.insert(0, 'Fold',       range(1,cv_cnt+1))
        train_df.insert(0, 'Seed',       i)
        train_df.insert(0, 'NeuronsCnt', elm.n)
        train_df.insert(0, 'L2',         elm.l2_param)
        big_train_multifold.append(train_df)

        test_df = pd.DataFrame(test_cv)
        test_df.insert(0, 'Fold',       range(1,cv_cnt+1))
        test_df.insert(0, 'Seed',       i)
        test_df.insert(0, 'NeuronsCnt', elm.n)
        test_df.insert(0, 'L2',         elm.l2_param)
        big_test_multifold.append(test_df)

    # Append Train DF
    train_metrics_df = pd.DataFrame(train_res)
    train_metrics_df = pd.concat(
        (pd.DataFrame([
            f' ELM with {int(elm_best_neuroncnt) } neurons',
            f'RELM with {int(elm_best_neuroncnt) } neurons, L2: {relm_best_l2}'
        ], columns=['Description']), train_metrics_df
        ), axis=1)
    big_train_res.append(train_metrics_df)

    # Append Test Df
    metrics_df = pd.DataFrame(res)
    metrics_df = pd.concat(
        (pd.DataFrame([
            f' ELM with {int(elm_best_neuroncnt) } neurons',
            f'RELM with {int(elm_best_neuroncnt) } neurons, L2: {relm_best_l2}'
        ], columns=['Description']), metrics_df
        ), axis=1)
    big_res.append(metrics_df)
```

<h3>Place CV Multi Seed to CSV</h3>

```python
big_test_multi_df = pd.concat(big_test_multifold)
elm_multi_df  = big_test_multi_df.loc[
    (big_test_multi_df['L2'] == 0)
].drop(['NeuronsCnt', 'L2'], axis=1)
relm_multi_df = big_test_multi_df.loc[
    (big_test_multi_df['L2'] != 0) &
    (big_test_multi_df['NeuronsCnt'] == big_test_multi_df['NeuronsCnt'].max())
].drop(['NeuronsCnt', 'L2'], axis=1)

relm_sub_multi_df = big_test_multi_df.loc[
    (big_test_multi_df['L2'] != 0) &
    (big_test_multi_df['NeuronsCnt'] == big_test_multi_df['NeuronsCnt'].min())
].drop(['NeuronsCnt', 'L2'], axis=1)
```
