```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from matplotlib.ticker import MultipleLocator
# scikit-learn == 1.5.2
from sklearn.model_selection import train_test_split
import time

from ELM import ELMclf
from preprocess import preprocess, get_data

from utils import cross_val_score, get_metrics
```

```python
split_seed = 42
```

```python
path_to_data = '../data/'
```

```python
# Acquire preprocessed dataset
df = get_data(f'{path_to_data}thyroid_diff.csv')
df = preprocess(df, do_onehot=True, do_scale=True)

X = df.drop('Recurred_Yes', axis=1)
Y = df['Recurred_Yes']
x_train,     x_test,     y_train,     y_test     = train_test_split(X, Y, test_size=0.2, random_state=split_seed, stratify=Y)
x_train_sub, x_validate, y_train_sub, y_validate = train_test_split(x_train, y_train, test_size=0.25, random_state=split_seed, stratify=y_train)
```

```python
###
### Only Neuron cnt
def multimodel_cross_val_score(estimator, X_param, y=None, scoring=None, cv=None, shuffle=False, stratified=True, random_seed_range=range(0, 100)):
    ls = []
    for i_inner in random_seed_range:
        estimator.random_state = i_inner
        score_i = cross_val_score(estimator, X_param, y, scoring, cv, shuffle, stratified, random_state=i_inner)
        ls.append(score_i)

    return np.array(ls)


def generate_with_parameters(neurons_cnt=range(1,150), l2_params=np.arange(0.,1.), model_seeds=range(42,43)):
    texts = []
    neuron_for_pd = []
    l2_for_pd = []
    model_seed_for_pd = []

    acc_avgs  = []; acc_maxs  = []; acc_mins  = []; acc_stds  = []
    prec_avgs = []; prec_maxs = []; prec_mins = []; prec_stds = []
    rec_avgs  = []; rec_maxs  = []; rec_mins  = []; rec_stds  = []
    f1_avgs   = []; f1_maxs   = []; f1_mins   = []; f1_stds   = []
    f2_avgs   = []; f2_maxs   = []; f2_mins   = []; f2_stds   = []
    spec_avgs = []; spec_maxs = []; spec_mins = []; spec_stds = []
    npv_avgs  = []; npv_maxs  = []; npv_mins  = []; npv_stds  = []
    mcc_avgs  = []; mcc_maxs  = []; mcc_mins  = []; mcc_stds  = []
    generic_avgs = []; generic_maxs = []; generic_mins = []; generic_stds = []

    scoring_func = get_metrics
    for i_inner in neurons_cnt:
        start = time.time()
        for l2 in l2_params:
            l2_start = time.time()
            ls = np.array(multimodel_cross_val_score(ELMclf(i_inner, l2_param=l2), x_train.to_numpy(), y_train.to_numpy(), cv=5, scoring=scoring_func, random_seed_range=model_seeds))

            if l2 == 0: texts.append(f' ELM of {i_inner} neurons')
            else:       texts.append(f'RELM of {i_inner} neurons & l2_param of {l2}')

            # if not get metrics, just return one
            if scoring_func.__name__ != 'get_metrics':
                for _, _ in enumerate(model_seeds):
                    neuron_for_pd.append(i_inner)
                    l2_for_pd.append(l2)

                    generic_avgs.append(ls.mean())
                    generic_maxs.append(ls.max())
                    generic_mins.append(ls.min())
                    generic_stds.append(ls.std())
            else:
                # decompose the list of dicts
                for idx, m in enumerate(model_seeds):
                    neuron_for_pd.append(i_inner)
                    l2_for_pd.append(l2)
                    model_seed_for_pd.append(m)
                    accuracies = np.array([metrics['accuracy']    for metrics in ls[idx]])
                    precisions = np.array([metrics['precision']   for metrics in ls[idx]])
                    recalls    = np.array([metrics['recall']      for metrics in ls[idx]])
                    f1s        = np.array([metrics['f1']          for metrics in ls[idx]])
                    f2s        = np.array([metrics['f2']          for metrics in ls[idx]])
                    specs      = np.array([metrics['specificity'] for metrics in ls[idx]])
                    npvs       = np.array([metrics['npv']         for metrics in ls[idx]])
                    mccs       = np.array([metrics['mcc']         for metrics in ls[idx]])


                    acc_avgs.append(accuracies.mean())
                    acc_maxs.append(accuracies.max())
                    acc_mins.append(accuracies.min())
                    acc_stds.append(accuracies.std())

                    prec_avgs.append(precisions.mean())
                    prec_maxs.append(precisions.max())
                    prec_mins.append(precisions.min())
                    prec_stds.append(precisions.std())

                    rec_avgs.append(recalls.mean())
                    rec_maxs.append(recalls.max())
                    rec_mins.append(recalls.min())
                    rec_stds.append(recalls.std())

                    f1_avgs.append(f1s.mean())
                    f1_maxs.append(f1s.max())
                    f1_mins.append(f1s.min())
                    f1_stds.append(f1s.std())

                    f2_avgs.append(f2s.mean())
                    f2_maxs.append(f2s.max())
                    f2_mins.append(f2s.min())
                    f2_stds.append(f2s.std())

                    spec_avgs.append(specs.mean())
                    spec_maxs.append(specs.max())
                    spec_mins.append(specs.min())
                    spec_stds.append(specs.std())

                    npv_avgs.append(npvs.mean())
                    npv_maxs.append(npvs.max())
                    npv_mins.append(npvs.min())
                    npv_stds.append(npvs.std())

                    mcc_avgs.append(mccs.mean())
                    mcc_maxs.append(mccs.max())
                    mcc_mins.append(mccs.min())
                    mcc_stds.append(mccs.std())


            print(f"{time.time() - l2_start:.4f} s: Neuron {i_inner}, L2 {l2}")
        print(f"\t\t\tNeuron {i_inner}: {time.time() - start:.2f} seconds")

    if scoring_func.__name__ != 'get_metrics':
        df_param = pd.DataFrame(np.transpose(np.array(
            [
             texts,
             neuron_for_pd,
             l2_for_pd,
             np.array(generic_avgs),
             np.array(generic_maxs),
             np.array(generic_mins),
             np.array(generic_stds),]
        )), columns=['NeuronsCnt', 'L2', 'Avg', 'Max', 'Min', 'Std'])
    else:
        df_param = pd.DataFrame(np.transpose(np.array(
        [
         # texts,
         model_seed_for_pd,
         neuron_for_pd,
         l2_for_pd,
         np.array(acc_avgs),  np.array(acc_maxs),  np.array(acc_mins),  np.array(acc_stds),
         np.array(prec_avgs), np.array(prec_maxs), np.array(prec_mins), np.array(prec_stds),
         np.array(rec_avgs),  np.array(rec_maxs),  np.array(rec_mins),  np.array(rec_stds),
         np.array(f1_avgs),   np.array(f1_maxs),   np.array(f1_mins),   np.array(f1_stds),
         np.array(f2_avgs),   np.array(f2_maxs),   np.array(f2_mins),   np.array(f2_stds),
         np.array(spec_avgs), np.array(spec_maxs), np.array(spec_mins), np.array(spec_stds),
         np.array(npv_avgs),  np.array(npv_maxs),  np.array(npv_mins),  np.array(npv_stds),
         np.array(mcc_avgs),  np.array(mcc_maxs),  np.array(mcc_mins),  np.array(mcc_stds),
         ]
        )), columns=[
            # 'Desc',
            'ModelSeed',
            'NeuronsCnt', 'L2',
            'Acc_Avg',  'Acc_Max',  'Acc_Min',  'Acc_Std',
            'Prec_Avg', 'Prec_Max', 'Prec_Min', 'Prec_Std',
            'Rec_Avg',  'Rec_Max',  'Rec_Min',  'Rec_Std',
            'F1_Avg',   'F1_Max',   'F1_Min',   'F1_Std',
            'F2_Avg',   'F2_Max',   'F2_Min',   'F2_Std',
            'Spec_Avg', 'Spec_Max', 'Spec_Min', 'Spec_Std',
            'NPV_Avg',  'NPV_Max',  'NPV_Min',  'NPV_Std',
            'MCC_Avg',  'MCC_Max',  'MCC_Min',  'MCC_Std',
        ])
    return df_param
```

```python
def plot_with(df_param, title=None, use_scatter=True, ylabel='F1', showwhichcol=None, xaxis_scale=None, yaxis_scale=None, y_lim=None, show_std=True, isint=False, select_highest=True, color='cornflowerblue', ax=None, bbox=(0.5, -0.24)):
    # """
    # Given dataframe with 5 columns, plot them
    # :param df_param:
    # :param title:
    # :param use_scatter:
    # :return:
    # """
    x_axis, col_avg, col_max, col_min, col_std = df_param.columns
    if use_scatter:
        axs = df_param.plot.scatter(x_axis, col_avg, color=color, title=title, ax=ax)
        df_param.plot.scatter(x_axis, col_max, color='cyan', ax=axs)
        df_param.plot.scatter(x_axis, col_min, color='orange', ax=axs)
    else:
        axs = df_param.plot.line(x_axis, col_avg, color=color, title=title, ax=ax)
        # df.plot.line(x_axis, col_max, color='cyan', ax=axs, alpha=0.4)
        # df.plot.line(x_axis, col_min, color='orange', ax=axs, alpha=0.4)
    # df.plot.line(x_axis, col_std, color='green', ax=axs)
    axs.margins(x=0.01,y=0.01)
    if xaxis_scale is not None:
        if xaxis_scale[0] is not None: axs.xaxis.set_major_locator(MultipleLocator(xaxis_scale[0]))
        if xaxis_scale[1] is not None: axs.xaxis.set_minor_locator(MultipleLocator(xaxis_scale[1]))
    if yaxis_scale is not None:
        if yaxis_scale[0] is not None: axs.yaxis.set_major_locator(MultipleLocator(yaxis_scale[0]))
        if yaxis_scale[1] is not None: axs.yaxis.set_minor_locator(MultipleLocator(yaxis_scale[1]))

    if y_lim is not None:
        axs.set_ylim(*y_lim)
    axs.grid(which='major', linestyle=':', linewidth=0.5)
    axs.spines['top'].set_visible(False)
    axs.spines['right'].set_visible(False)
    axs.set_xlabel(axs.get_xlabel(), fontweight='bold')
    axs.set_ylabel(ylabel, fontweight='bold')

    if show_std:
        plt.fill_between(df_param[x_axis], df_param[col_avg] - df_param[col_std], df_param[col_avg] + df_param[col_std], alpha=0.25, label='Std') # std area
    # df.plot.line('NeuronsCnt', 'Avg/Std', color='green')
    # axs.plot(df['NeuronsCnt'], df['Max']-df['Min'], color='green')


    print(f'Best {col_max} performance: ')
    max_best = df_param.loc[df_param[col_max].idxmax()]
    print(max_best)
    if showwhichcol == 2:
        plt.axvline(max_best[x_axis], color=color, linestyle='dotted', label=f'Best {x_axis}: {int(max_best[x_axis]) if isint else max_best[x_axis]}', alpha=0.8)
        # plt.axvline(max_best[x_axis], color='cyan', linestyle='dotted', label='Best Max', alpha=0.8)

    print(f'Best {col_min} performance: ')
    min_best = df_param.loc[df_param[col_min].idxmax()]
    print(min_best)
    if showwhichcol == 3:
        plt.axvline(min_best[x_axis], color=color, linestyle='dotted', label=f'Best {x_axis}: {int(min_best[x_axis]) if isint else min_best[x_axis]}', alpha=0.8)
        # plt.axvline(min_best[x_axis], color='orange', linestyle='dashed', label='Best Min', alpha=0.5)

    print(f'Best {col_avg} Performance: ')
    if select_highest:
        avg_best = df_param.loc[df_param[col_avg].idxmax()]
    else:
        avg_best = df_param.loc[df_param[col_avg].idxmin()]
    print(avg_best)
    if showwhichcol == 1:
        plt.axvline(avg_best[x_axis], color=color, linestyle='dotted', label=f'Best {x_axis}: {int(avg_best[x_axis]) if isint else avg_best[x_axis]}', alpha=0.8)

    print(f'Best {col_std} Performance: ')
    std_best = df_param.loc[df_param.loc[df_param[x_axis] != 1, col_std].idxmin()]
    print(std_best)
    if showwhichcol == 4:
        plt.axvline(std_best[x_axis], color=color, linestyle='dotted', label=f'Best {x_axis}: {int(std_best[x_axis]) if isint else std_best[x_axis]}', alpha=0.8)
        # plt.axvline(std_best[x_axis], color='green', linestyle='dotted', label='Best Std', alpha=0.8)

    ncol = len(axs.get_legend_handles_labels()[1])
    axs.legend(loc='lower center', bbox_to_anchor=bbox, ncol=ncol, frameon=False, fontsize=10)
    return axs

```

```python
n_low, n_high = 2, 100
neurons = range(n_low, n_high+1)

l2_low, l2_high, l2_inc = 0., 1., 0.04
l2s = np.arange(l2_low, l2_high, l2_inc)

seed_low, seed_high, seed_inc = 0, 100, 2
seeds = range(seed_low, seed_high, seed_inc)
hidden_neuron_df = generate_with_parameters(neurons, l2s, seeds)
```

```python
fname = f'{path_to_data}node{n_low}_{n_high}__seed{seed_low}_{seed_high}_{seed_inc}__l0_1_{str(l2_inc).replace('.','')}.csv'
hidden_neuron_df.to_csv(fname)
```
