```python
import os.path

from matplotlib.ticker import MultipleLocator
# from ucimlrepo import fetch_ucirepo # fetch dataset
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from utils import subplot_txt
```

```python
# This is an alternative to CSV
# dtc_dat = fetch_ucirepo(id=915)

# Converts csv dataset to pandas
fname = 'data/thyroid_diff.csv'
if os.path.isfile(fname):
    df = pd.read_csv(fname)
else:
    raise FileNotFoundError(fname)

del fname
```

```python
def corr_df(df, col_name, target, percentage=False):
    df_freq = None
    df_percent = None
    if target is None:
        df_freq = df[col_name].value_counts()
        df_percent = df_freq / df.shape[0]
        if percentage:
            return df_percent
        else:
            return df_freq
    for stage in df[col_name].unique():
        df_temp = (df.loc[df[col_name]==stage, [col_name, target]]
                   .value_counts() # Get frequency
                   .to_frame()    # Series -> DataFrame
                   .reset_index() # Was multi-index
                   .pivot(index=col_name, columns=target, values='count')) # Set index to recurred
        freq = df_temp.sum(axis=1)[stage]

        # temp2 gets the percentage version
        df_temp2 = df_temp.divide(freq)
        if percentage and df_percent is None:
            df_percent = df_temp2
        else:
            df_percent = pd.concat([df_percent, df_temp2], axis=0)

        if not percentage and df_freq is None:
            df_freq = df_temp
        else:
            df_freq = pd.concat([df_freq, df_temp], axis=0)

    df_freq.fillna(0, inplace=True)
    df_percent.fillna(0, inplace=True)
    if percentage:
        return df_percent
    else:
        return df_freq


def category_plot(
        corr, df,
        colx, coly,
        title=None, ax=None, reindex=None, reindex_y=None,
        barh=False,
        xticks_label=None, yticks_label=None,
        axis_size=None, title_size=None, xticks_size=10, barlabel_size=None, legend_size=None,
        yaxis_scale=None,
        y_lim=None,
        subplotname=None, color=None, cmap='seismic', use_percentage=False, use_stack=False):

    if use_percentage:
        df_plot = corr(df, colx, coly, percentage=True)
    else:
        df_plot = corr(df, colx, coly, percentage=False)

    if reindex is not None:
        df_plot = df_plot.reindex(reindex, axis=0)
    if reindex_y is not None:
        df_plot = df_plot.reindex(reindex_y, axis=1)

    bar = 'barh' if barh else 'bar'
    if color is not None:
        ax = df_plot.plot(kind=bar, color=color, ax=ax, stacked=use_stack)
    else:
        ax = df_plot.plot(kind=bar, colormap=cmap, ax=ax, stacked=use_stack)
    ax.set_title(title, pad=20)
    if title_size is not None: ax.set_title(title, fontsize=title_size, pad=20)

    ylabel = 'Percentage' if use_percentage else 'Count'
    ax.set_xlabel(colx, fontweight='bold', fontsize=axis_size)
    ax.set_ylabel(ylabel, fontweight='bold', fontsize=axis_size)
    if xticks_label is not None:
        ax.set_xticks(ticks=range(len(xticks_label)), labels=xticks_label)
        ax.tick_params(axis='x', rotation=0, labelsize=xticks_size)
    else:
        ax.tick_params(axis='x', rotation=0, labelsize=xticks_size)

    if yticks_label is not None:
        ax.set_yticks(ticks=range(len(yticks_label)), labels=yticks_label, rotation=0)
    else:
        ax.tick_params(axis='y', rotation=0)

    if yaxis_scale is not None:
        if yaxis_scale[0] is not None: ax.yaxis.set_major_locator(MultipleLocator(yaxis_scale[0]))
        if yaxis_scale[1] is not None: ax.yaxis.set_minor_locator(MultipleLocator(yaxis_scale[1]))
    if y_lim is not None:
        ax.set_ylim(*y_lim)

    label_type = 'center' if use_stack else 'edge'
    if use_percentage:
        fmt = lambda x: f'{x:.2f}' if x > 0 else ''
    else:
        fmt = lambda x: f'{x:.0f}' if x > 0 else ''
    colors = list(dict.fromkeys([p.get_facecolor() for p in ax.patches]))
    for c,color in zip(ax.containers, colors):
        # https://learn.microsoft.com/en-us/answers/questions/1497116/how-to-find-if-a-given-color-is-dark-or-light
        # thank you
        is_bright = (color[0] * 0.2126) + (color[1] * 0.7152) + (color[2] * 0.0722) > 0.5
        label_color = 'black' if is_bright or not use_stack else 'white'
        ax.bar_label(c, padding=5, label_type=label_type, fmt=fmt, color=label_color, size=barlabel_size)

    ax.minorticks_on()
    ax.tick_params(axis='x', which='minor',bottom=False)
    ax.grid(axis='y', which='major', linestyle='--', linewidth=0.5, alpha=0.4)
    ax.grid(axis='y', which='minor', linestyle=':', linewidth=0.2, alpha=0.4)
    # ax.tick_params(axis='x', which='major',bottom=False)


    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

    if subplotname is not None:
        subplot_txt(subplotname, ax=ax, x=-0.12, y=1.13)

    ncol = 1 if df_plot.ndim == 1 else df_plot.shape[1]
    # ax.legend(frameon=False)
    ax.legend(loc='lower center', bbox_to_anchor=(0.5, -0.24), ncol=ncol, frameon=False, fontsize=legend_size)
    # df_plot_percent.plot(kind='bar', stacked=True,colormap='seismic')
    # ax.set_title('Recurrence based on Stage - Percentage')
    return ax

def plot_hist(
        df, col,
        title=None, ax=None,
        bins=10,
        xaxis_scale=None,
        yaxis_scale=None,
        x_lim=None, y_lim=None,
        axis_size=None, title_size=None, xticks_size=10, barlabel_size=None,
        subplotname=None, color=None, cmap='seismic', use_percentage=False):

    df_plot = df.copy()
    n = df.shape[0]
    if ax is None:
        _, ax = plt.subplots()

    if use_percentage:
        weights = np.ones(n) / n
    else:
        weights = None
    if color is not None:
        n, bins, patches = ax.hist(df_plot[col], bins=bins, color=color, weights=weights)
    else:
        n, bins, patches = ax.hist(df_plot[col], bins=bins, weights=weights)

        # get centre of each bin
        bin_centers = 0.5 * (bins[:-1] + bins[1:])
        scaled_bin_centers = (bin_centers - min(bin_centers)) / max(bin_centers)
        cm = plt.colormaps[cmap]
        for c, p in zip(scaled_bin_centers, patches):
            plt.setp(p, 'facecolor', cm(c))

    ax.set_title(title, fontsize=title_size)
    ylabel = 'Percentage' if use_percentage else 'Count'
    ax.set_xlabel(col, fontweight='bold', fontsize=axis_size)
    ax.set_ylabel(ylabel, fontweight='bold', fontsize=axis_size)


    if xaxis_scale is not None:
        if xaxis_scale[0] is not None: ax.xaxis.set_major_locator(MultipleLocator(xaxis_scale[0]))
        if xaxis_scale[1] is not None: ax.xaxis.set_minor_locator(MultipleLocator(xaxis_scale[1]))
    if yaxis_scale is not None:
        if yaxis_scale[0] is not None: ax.yaxis.set_major_locator(MultipleLocator(yaxis_scale[0]))
        if yaxis_scale[1] is not None: ax.yaxis.set_minor_locator(MultipleLocator(yaxis_scale[1]))
    if x_lim is not None:
        ax.set_y_lim(*x_lim)
    if y_lim is not None:
        ax.set_ylim(*y_lim)

    ax.minorticks_on()
    ax.tick_params(axis='x', which='minor',bottom=False , labelsize=xticks_size)
    ax.grid(False, axis='x')
    ax.grid(axis='y', which='major', linestyle='--', linewidth=0.5, alpha=0.4)
    ax.grid(axis='y', which='minor', linestyle=':', linewidth=0.2, alpha=0.4)

    fmt = '%.2f' if use_percentage else ''
    ax.bar_label(patches, padding=5, fmt=fmt, rotation=0, size=barlabel_size)

    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

    if subplotname is not None:
        subplot_txt(subplotname, ax=ax, x=-0.12, y=1.07)
```
