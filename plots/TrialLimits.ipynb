```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
def get_agg(df, single_fold=0, first_avg_folds=True):
    # use fold number to get avg and std of the folds
    # first_avg_fold: compare stability of ABC between seeds
    # first_avg_seed: compare model generalisability
    if single_fold > 0:
        fold_df = df.loc[df['Fold'] == single_fold]
        avg_sub_std_df = fold_df
        avg_inner_df = fold_df
        std_seed_df = fold_df.std()
    elif first_avg_folds:
        group = df.groupby('Seed', as_index=False)
        avg_inner_df = group.mean()
        std_seed_df = group.std()
        avg_sub_std_df = avg_inner_df.drop(['Seed'], axis=1) - std_seed_df.drop(['Seed'], axis=1)  # In each iters find lower bound per fold
    else:
        group = df.groupby('Fold', as_index=False)
        avg_inner_df = group.mean()
        std_fold_df = group.std()
        avg_sub_std_df = avg_inner_df - std_fold_df
    avg_df = avg_sub_std_df.mean() # avg of models
    std_df = avg_inner_df.std()  # std of that
    return avg_df, std_df#, avg_sub_std_df
```

```python
def get_agg_from(solution_cnt, max_iter, trial_limit, use_train=False, single_fold=0, first_avg_folds=True):
    fname = f'../data/abc_scores_final_n11_sol{solution_cnt}_iter{max_iter}_lim{trial_limit}.csv'

    scores_df = pd.read_csv(fname, index_col=0)
    train = scores_df.loc[scores_df['predicted_data'] == 'train_set'].drop('predicted_data', axis=1)
    test = scores_df.loc[scores_df['predicted_data'] == 'test_set'].drop('predicted_data', axis=1)

    if use_train: avg_df, std_df = get_agg(train, single_fold=single_fold, first_avg_folds=first_avg_folds)
    else:         avg_df, std_df = get_agg(test, single_fold=single_fold, first_avg_folds=first_avg_folds)
    # avg_df.insert(0, 'solution_cnt',solution_cnt)
    # avg_df.insert(1, 'max_iter',    max_iter)
    # avg_df.insert(2, 'trial_limit', trial_limit)
    series = pd.Series([solution_cnt, max_iter, trial_limit], index=['solution_cnt', 'max_iter', 'trial_limit'])
    avg_df = pd.concat([series, avg_df])
    std_df = pd.concat([series, std_df])
    return avg_df, std_df
```

```python
def get_box_plot(solution_cnt, max_iter, trial_limit, single_fold=0, metric_name='f1', first_avg_folds=True):
    fname = f'../data/abc_scores_final_n11_sol{solution_cnt}_iter{max_iter}_lim{trial_limit}.csv'

    scores_df = pd.read_csv(fname, index_col=0)
    train = scores_df.loc[scores_df['predicted_data'] == 'train_set'].drop('predicted_data', axis=1)
    test = scores_df.loc[scores_df['predicted_data'] == 'test_set'].drop('predicted_data', axis=1)

    if single_fold > 0:
        fold_df = test.loc[test['Fold'] == single_fold]
        avg_sub_std_df = fold_df
        std_seed_df = fold_df.std()
    else:
        if not first_avg_folds:
            group = test.groupby('Fold', as_index=False)
            avg_fold_df = group.mean()
            std_fold_df = group.std()
            avg_sub_std_df = avg_fold_df - std_fold_df
        else:
            group = test.groupby('Seed', as_index=False)
            avg_seed_df = group.mean()
            std_seed_df = group.std()
            avg_sub_std_df = avg_seed_df - std_seed_df
            # avg_sub_std_df = avg_seed_df.drop(['Seed'], axis=1)# - std_seed_df.drop(['Seed'], axis=1)  # In each iters find lower bound per fold

    return avg_sub_std_df[metric_name].rename(trial_limit/solution_cnt)
```

```python
solu_cnt   = 15
max_iter   = 300
trial_limit = [15,30,45]
param_list = [
    {
        'solution_cnt': solu_cnt,
        'max_iter': max_iter,
        'trial_limit': tl
    } for tl in trial_limit
]
param_list = [
    {
        'solution_cnt': solu_cnt,
        'max_iter': max_iter,
        'trial_limit': trial_limit
    } for trial_limit in trial_limit
]
```

```python
param_list
```

```python
fold = 0
avg_on_fold_first = True
train_avg_std_tup_list = [get_agg_from(**param, use_train=True, single_fold=fold, first_avg_folds=avg_on_fold_first) for param in param_list]
test_avg_std_tup_list  = [get_agg_from(**param, use_train=False, single_fold=fold, first_avg_folds=avg_on_fold_first) for param in param_list]
```

```python
train_avgs = [tup[0] for tup in train_avg_std_tup_list]
train_stds = [tup[1] for tup in train_avg_std_tup_list]
test_avgs  = [tup[0] for tup in test_avg_std_tup_list]
test_stds  = [tup[1] for tup in test_avg_std_tup_list]
```

```python
train_avgs_df = pd.concat(train_avgs,axis=1)
train_stds_df = pd.concat(train_stds,axis=1)
test_avgs_df  = pd.concat(test_avgs,axis=1)
test_stds_df  = pd.concat(test_stds,axis=1)
```

```python
for_plot = pd.concat([get_box_plot(solu_cnt,max_iter,t,0, first_avg_folds=True) for t in trial_limit],axis=1)
ax = sns.boxplot(for_plot, color='cornflowerblue')
sns.swarmplot(for_plot, color='white', edgecolor='black', linewidth=0.2, ax=ax)
```
