```python
import pandas as pd
from matplotlib import pyplot as plt, gridspec

from utils import train_test_graph_multiseed, train_test_graph_multiseed_2, subplot_txt
```

```python
def get_trace_df(solu_cnt, mi, tl, employed_legacy=1, onlooker_legacy=1):
    fname = f'../data/abc_final_n{int(employed_legacy)}{int(onlooker_legacy)}_sol{solu_cnt}_iter{mi}_lim{tl}.csv'
    return pd.read_csv(fname, index_col=0)
```

```python
def plots_on(use_test=True, first_avg_folds=False, use_which=None):
    start = 97
    plots_name = [f'({chr(i)})' for i in range(start, start+26)]
    for i, (df, (solu, mi, tl)) in enumerate(zip(dfs, param_lists)):
        print(f'Max Iter: {mi}')
        fig, ax_up = train_test_graph_multiseed(df, use_test, first_avg_folds, use_which=use_which, ylabel='F1', title=f'Train-Validation of {solu} Solutions & {mi} Iterations of 5 Splits', xaxis_scale=(50,None), y_lim=(0.88,0.96), no_scoutcnt=True, legend_size=11, label_size=13, title_size=14, tick_size=11)
        txt = subplot_txt(plots_name[i], ax_up, y=1.125)
        txt.set_in_layout(False)
```

```python
solution_cnts = [ 15, 15, 15, 15,  15]
max_iters     = [200,300,500,750,1000]
trial_limit = list(map(lambda x: x*2, solution_cnts))
params = zip(solution_cnts, max_iters, trial_limit)
param_lists = list(params)
dfs = [get_trace_df(solu, mi, tl)
       for solu, mi, tl in zip(solution_cnts, max_iters, trial_limit)]
```

```python
plots_on(use_which=['Validation', 'Test'])
```

```python
plots_on(first_avg_folds=True)
```
