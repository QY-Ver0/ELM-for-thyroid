```python
import os
import seaborn as sns
import pandas as pd
from utils import subplot_txt
```

```python
# average the scores so there is a single value to describe them (and also returns the standard deviation for reference)
def get_agg(df, single_fold=0, first_avg_folds=True):
    # use fold number to get avg and std of the folds
    # first_avg_fold: compare stability of ABC between seeds
    # first_avg_seed: compare model generalisability
    if single_fold > 0:
        fold_df = df.loc[df['Fold'] == single_fold]
        avg_sub_std_df = fold_df
        avg_inner_df = fold_df
        std_seed_df = fold_df.std()
    elif first_avg_folds:
        group = df.groupby('Seed', as_index=False)
        avg_inner_df = group.mean()
        std_seed_df = group.std()
        avg_sub_std_df = avg_inner_df.drop(['Seed'], axis=1) - std_seed_df.drop(['Seed'], axis=1)  # In each iters find lower bound per fold
    else:
        group = df.groupby('Fold', as_index=False)
        avg_inner_df = group.mean()
        std_fold_df = group.std()
        avg_sub_std_df = avg_inner_df - std_fold_df
    avg_df = avg_sub_std_df.mean() # avg of models
    std_df = avg_inner_df.std()  # std of that
    return avg_df, std_df#, avg_sub_std_df

def get_agg_from(solution_cnt, max_iter, trial_limit, employed_legacy=None, onlooker_legacy=None, use_train=False, single_fold=0, first_avg_folds=True):
    fname = f'../data/abc_scores_final_n{int(employed_legacy)}{int(onlooker_legacy)}_sol{solution_cnt}_iter{max_iter}_lim{trial_limit}.csv'
    scores_df = pd.read_csv(fname, index_col=0)
    train = scores_df.loc[scores_df['predicted_data'] == 'train_set'].drop('predicted_data', axis=1)
    test = scores_df.loc[scores_df['predicted_data'] == 'test_set'].drop('predicted_data', axis=1)

    if use_train: avg_df, std_df = get_agg(train, single_fold=single_fold, first_avg_folds=first_avg_folds)
    else:         avg_df, std_df = get_agg(test, single_fold=single_fold, first_avg_folds=first_avg_folds)

    series = pd.Series([solution_cnt, max_iter, trial_limit], index=['solution_cnt', 'max_iter', 'trial_limit'])
    avg_df = pd.concat([series, avg_df])
    std_df = pd.concat([series, std_df])
    return avg_df, std_df


def get_missing_params(s_list, mi_list, mult, employed_legacy, onlooker_legacy):
    all_case = [{
        'solution_cnt': solu_cnt,
        'max_iter': max_i,
        'trial_limit': solu_cnt * mult,
        'employed_legacy': employed_legacy,
        'onlooker_legacy': onlooker_legacy,
    } for solu_cnt in s_list for max_i in mi_list]

    return zip(all_case, [os.path.exists(f'../data/abc_scores_final_n{int(employed_legacy)}{int(onlooker_legacy)}_sol{solu_cnt}_iter{max_i}_lim{solu_cnt*mult}.csv') for solu_cnt in s_list for max_i in mi_list])
```

```python
# a function to save their validation score with the trace df
def get_agg_vali_score(s,mi,tl,employed_use_legacy, onlooker_use_legacy, single_fold=0, first_avg_folds=True):
    fname = f'../data/abc_final_n{int(employed_use_legacy)}{int(onlooker_use_legacy)}_sol{s}_iter{mi}_lim{tl}.csv'
    df = pd.read_csv(fname, index_col=0)
    group = df.groupby(['Seed', 'Fold'], as_index=False)
    last_rows = group.last()
    vali_avg_df, vali_std_df = get_agg(last_rows, single_fold, first_avg_folds)

    info = pd.Series([s,mi,tl], index=['solution_cnt', 'max_iter', 'trial_limit'])
    return (pd.concat([info,vali_avg_df]),
            pd.concat([info,vali_std_df]))

def get_all_vali_score(s_list, mi_list, mult, employed_legacy, onlooker_legacy, single_fold=0, first_avg_folds=True):
    all_case = [{
        's' : solu_cnt,
        'mi': max_i,
        'tl': solu_cnt * mult,
        'employed_use_legacy': employed_legacy,
        'onlooker_use_legacy': onlooker_legacy,
        'single_fold'    : single_fold,
        'first_avg_folds': first_avg_folds
    } for solu_cnt in s_list for max_i in mi_list]
    return (pd.concat([get_agg_vali_score(**case)[0] for case in all_case], axis=1).T,
            pd.concat([get_agg_vali_score(**case)[1] for case in all_case], axis=1).T)
```

```python
all_s      = [15, 20, 30, 45, 60, 75]
all_mi     = [200,300,500,750,1000]
employed_legacy = True
onlooker_legacy = True

path_exists = list(get_missing_params(all_s,all_mi, 2, employed_legacy, onlooker_legacy))
available = [tup for tup in path_exists if tup[1]]
```

```python
param_list = [tup[0] for tup in available]
```

```python
fold = 0
first_avg_folds = False
train_avg_std_tup_list = [get_agg_from(**param, use_train=True, single_fold=fold, first_avg_folds=first_avg_folds) for param in param_list]
test_avg_std_tup_list  = [get_agg_from(**param, use_train=False, single_fold=fold, first_avg_folds=first_avg_folds) for param in param_list]
```

```python
vali_avgs_df, vali_stds_df = get_all_vali_score(all_s, all_mi, 2, employed_legacy, onlooker_legacy, single_fold=fold, first_avg_folds=first_avg_folds)
vali_avgs_df.drop(['Train', 'Test', 'ScoutCall'], axis=1, inplace=True)
vali_stds_df.drop(['Train', 'Test', 'ScoutCall'], axis=1, inplace=True)
vali_avgs_df.rename({'Validation': 'f1'}, inplace=True, axis=1)
vali_stds_df.rename({'Validation': 'f1'}, inplace=True, axis=1)
```

```python
train_avgs = [tup[0] for tup in train_avg_std_tup_list]
train_stds = [tup[1] for tup in train_avg_std_tup_list]
test_avgs  = [tup[0] for tup in test_avg_std_tup_list]
test_stds  = [tup[1] for tup in test_avg_std_tup_list]
```

```python
train_avgs_df = pd.DataFrame(train_avgs)
test_avgs_df  = pd.DataFrame(test_avgs)
train_stds_df = pd.DataFrame(train_stds)
test_stds_df  = pd.DataFrame(test_stds)
```

```python
train_mat = train_avgs_df.pivot_table(index='solution_cnt', columns='max_iter', values='f1')
vali_mat  = vali_avgs_df.pivot_table(index='solution_cnt', columns='max_iter', values='f1')
test_mat  = test_avgs_df.pivot_table(index='solution_cnt', columns='max_iter', values='f1')
train_std_mat = train_stds_df.pivot_table(index='solution_cnt', columns='max_iter', values='f1')
vali_std_mat  = vali_stds_df.pivot_table(index='solution_cnt', columns='max_iter', values='f1')
test_std_mat  = test_stds_df.pivot_table(index='solution_cnt', columns='max_iter', values='f1')

train_mat = train_mat[::-1]
vali_mat  = vali_mat[::-1]
test_mat  = test_mat[::-1]
train_std_mat = train_std_mat[::-1]
vali_std_mat  = vali_std_mat[::-1]
test_std_mat  = test_std_mat[::-1]
```

```python
from scipy.ndimage import gaussian_filter
use_which = 0 # 0, 1, 2
gauss_param = {
    'mode':'nearest', 'sigma':0.7, 'radius':1
}
gauss_train = pd.DataFrame(gaussian_filter(train_mat, **gauss_param), index=train_mat.index, columns=train_mat.columns)
gauss_vali  = pd.DataFrame(gaussian_filter(vali_mat, **gauss_param), index=vali_mat.index, columns=vali_mat.columns)
gauss_test  = pd.DataFrame(gaussian_filter(test_mat, **gauss_param), index=test_mat.index, columns=test_mat.columns)

gauss_train_std = pd.DataFrame(gaussian_filter(train_std_mat, **gauss_param), index=train_mat.index, columns=train_mat.columns)
gauss_vali_std  = pd.DataFrame(gaussian_filter(vali_std_mat, **gauss_param), index=vali_mat.index, columns=vali_mat.columns)
gauss_test_std  = pd.DataFrame(gaussian_filter(test_std_mat, **gauss_param), index=test_mat.index, columns=test_mat.columns)

train_sub_std = train_mat - train_std_mat
vali_sub_std  = vali_mat  - vali_std_mat
test_sub_std  = test_mat  - test_std_mat

gauss_train_sub_std = pd.DataFrame(gaussian_filter(train_sub_std, **gauss_param), index=train_mat.index, columns=train_mat.columns)
gauss_vali_sub_std  = pd.DataFrame(gaussian_filter(vali_sub_std, **gauss_param), index=vali_mat.index, columns=vali_mat.columns)
gauss_test_sub_std  = pd.DataFrame(gaussian_filter(test_sub_std, **gauss_param), index=test_mat.index, columns=test_mat.columns)
```

```python
if use_which == 0:
    print('use_train')
    mat     = train_mat
    std_mat = train_std_mat
    gauss_mat = gauss_train
    gauss_std = gauss_train_std
    sub_std   = train_sub_std
    gauss_sub_std = gauss_train_sub_std
elif use_which == 1:
    print('use_vali')
    mat     = vali_mat
    std_mat = vali_std_mat
    gauss_mat = gauss_vali
    gauss_std = gauss_vali_std
    sub_std   = vali_sub_std
    gauss_sub_std = gauss_vali_sub_std
else:
    print('use_cv_vali')
    mat     = test_mat
    std_mat = test_std_mat
    gauss_mat = gauss_test
    gauss_std = gauss_test_std
    sub_std   = test_sub_std
    gauss_sub_std = gauss_test_sub_std

train_sub_cvvali = (train_sub_std - test_sub_std)
gauss_train_cvvali = pd.DataFrame(gaussian_filter(train_sub_cvvali, **gauss_param), index=train_mat.index, columns=train_mat.columns)
vali_sub_cvvali = (vali_sub_std - test_sub_std)
gauss_vali_cvvali = pd.DataFrame(gaussian_filter(vali_sub_cvvali, **gauss_param), index=vali_mat.index, columns=train_mat.columns)
```

```python
def heat(
        mat, title='', ax=None, fmt='.4f',
        cmap='coolwarm',
        vmin=None, vmax=None,
        ticks=None,
        label_size=None, title_size=None, annot_size=12):
    ax = sns.heatmap(mat, ax=ax, annot=True, fmt=fmt, cmap=cmap, vmin=vmin, vmax=vmax, cbar_kws={'ticks': ticks}, annot_kws={"size": annot_size})
    ax.set_title(title, fontsize=title_size)
    ax.set_xlabel(ax.get_xlabel(), fontweight='bold', fontsize=label_size)
    ax.set_ylabel(ax.get_ylabel(), fontweight='bold', fontsize=label_size)
    return ax
```

```python
train_vmin = 0.923
train_vmax = 0.933
label_size = 12
title_size = 14
```

```python
plots_name = ['a', 'b']
txt_func = lambda i: subplot_txt(f'({plots_name[i]})', None, -0.5, -0.32)
```

```python
train_ticks = [train_vmin + k*0.002 for k in range(0,6)]
heat(mat,title='Train F1 Performance', vmin=train_vmin, vmax=train_vmax, ticks=train_ticks, label_size=label_size, title_size=title_size)
txt_func(0)
```

```python
heat(gauss_mat, title='Smoothed Train F1 Performance', vmin=train_vmin, vmax=train_vmax, ticks=train_ticks, label_size=label_size, title_size=title_size)
txt_func(1)
```

```python
heat(gauss_test, title='Smoothed Validation F1 Performance', cmap='coolwarm', vmin=test_vmin, vmax=test_vmax, ticks=test_ticks, label_size=label_size, title_size=title_size)
txt_func(1)
```
