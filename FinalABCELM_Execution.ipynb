```python
import copy

import numpy as np
from joblib import Parallel, delayed

from ELM import ELMclf
import pandas as pd
from sklearn.model_selection import train_test_split

from preprocess import preprocess, get_data
from optimiser_main import Optimiser, ABCOptimiser, get_dims
from utils import f1, get_metrics
```

```python
# setting params
n            = 38
l2           = 0.36
def_rng_seed = 42
split_seed   = 42
global_rng   = np.random.default_rng(def_rng_seed)
defined_fn   = f1
cv           = 5

solution_cnt = 15
max_iter     = 300
trial_limit  = 30
```

```python
# Importing preprocessed dataset
df = get_data('data/thyroid_diff.csv')
df = preprocess(df, do_onehot=True, do_scale=True)

X, Y = df.drop('Recurred_Yes', axis=1), df['Recurred_Yes']
X_np, Y_np = np.asarray(X, dtype=np.float32), np.asarray(Y, dtype=np.float32)
x_train,     x_test,     y_train,     y_test     = train_test_split(X, Y, test_size=0.2, random_state=split_seed, stratify=Y)
x_train_sub, x_validate, y_train_sub, y_validate = train_test_split(x_train, y_train, test_size=0.25, random_state=split_seed, stratify=y_train)
```

```python
def multi_seed_abc(optimiser: Optimiser, x_train, y_train, x_test, y_test, args, seed_range=range(0,102,2), job=1):
    # val_size: validation size for split in train set of validation set
    optimiser_copy = copy.deepcopy(optimiser)

    res = Parallel(n_jobs=job,verbose=50)(delayed(indi_abc_loop)(
        optimiser_copy, i, x_train, y_train, x_test, y_test, args
    ) for i in seed_range)
    tt_dfs, train_scores, test_scores = tuple(map(list, zip(*res)))
    return tt_dfs, train_scores, test_scores

def indi_abc_loop(optimiser: ABCOptimiser, i, x_train, y_train, x_test, y_test, args):
    optimiser_copy = copy.deepcopy(optimiser)
    optimiser_copy.random_state = i
    best_param = optimiser_copy.fit(x_train, y_train, x_test, y_test, **args)
    # y_pred_test = optimiser_copy.predict(x_test)
    elm = ELMclf(optimiser_copy.models[0].n, optimiser_copy.models[0].l2_param, random_state=optimiser_copy.models[0].random_state)
    elm.fit(x_train, y_train, **best_param)
    y_pred_train = elm.predict(x_train)
    y_pred_test = elm.predict(x_test)
    train_score = get_metrics(y_train, y_pred_train)
    test_score = get_metrics(y_test, y_pred_test)

    train_res = optimiser_copy.train_res
    vali_res  = optimiser_copy.vali_res
    test_res  = optimiser_copy.test_res
    scouts_cnt = optimiser_copy.lim_reached_cnts
    tt_df = pd.DataFrame([train_res, vali_res, test_res, scouts_cnt],
                             index=['Train', 'Validation', 'Test', 'ScoutCall'])
    tt_df = tt_df.transpose()
    iters = range(1, tt_df.shape[0] + 1)
    tt_df.insert(0, 'Iters', iters)
    tt_df.insert(0, 'Fold', np.repeat(i + 1, tt_df.shape[0]))
    return tt_df, train_score, test_score

```

```python
# Getting initial solutions
elm = ELMclf(n, l2_param=l2, random_state=def_rng_seed)
elm.fit(x_train_sub.to_numpy(), y_train_sub.to_numpy())

popu_temp  = elm.return_args() # template
popu_temp.pop('l2_param')
dims = get_dims(popu_temp)
popu_range = {'weights_i': (-1,1), 'biases': (-1,1)}     # range
popu = Optimiser.population_generation(solution_cnt, popu_temp, popu_range, random_state=int(global_rng.uniform(0, 4294967295)))
abc_args = {
    'population':       popu,
    'population_limit': popu_range,
    'max_iter':         max_iter,
    'trial_limit':      trial_limit,
    'cv':               5,
    'cv_on_train':      False,
}
abc = ABCOptimiser(elm, fitness_fn=defined_fn, random_state=def_rng_seed)
abc.set_verbose(True)
```

```python
tt_dflist, train_scores, test_scores = multi_seed_abc(
            abc, x_train.to_numpy(), y_train.to_numpy(), x_test.to_numpy(), y_test.to_numpy(), abc_args,
            seed_range=range(0, 50), job=5)
```

```python
pd.concat(tt_dflist).to_csv(f'data/result_single_abc_trace_sol{solution_cnt}_iter{max_iter}_lim{trial_limit}.csv')
```
