```python
import numpy as np
import pandas as pd

# scikit-learn == 1.5.2
from sklearn.model_selection import train_test_split
import json

from ELM import ELMclf
from preprocess import preprocess, get_data

from utils import cross_val_score, get_metrics, format_metric_df
```

Setting Global Params

```python
split_seed = 42
r_seed     = 42
seed_range = range(0,50)
```

```python
path_to_data = '../../data/'
path_to_data_imp = '../../important_data/'
```

```python
# Acquire preprocessed dataset
df = get_data(f'{path_to_data}thyroid_diff.csv')
df = preprocess(df, do_onehot=True, do_scale=True)

X = df.drop('Recurred_Yes', axis=1)
Y = df['Recurred_Yes']
x_train,     x_test,     y_train,     y_test     = train_test_split(X, Y, test_size=0.2, random_state=split_seed, stratify=Y)
x_train_sub, x_validate, y_train_sub, y_validate = train_test_split(x_train, y_train, test_size=0.25, random_state=split_seed, stratify=y_train)
```

```python
# Import ELM and RELM configuration
with open(f'{path_to_data_imp}optimal_config.json', 'r') as file:
    elm_configs = json.load(file)

elm_best_neuroncnt = elm_configs['n']
relm_best_l2 = elm_configs['l2_param']
```

# <b>Run To Predict Test set Across Multiple Seeds</b>

```python
# multi seed check
big_train_res = []
big_res       = []
big_train_multifold = []
big_test_multifold  = []

elm1 = ELMclf(int(elm_best_neuroncnt), l2_param=0, random_state=r_seed)
elm2 = ELMclf(int(elm_best_neuroncnt), l2_param=relm_best_l2, random_state=r_seed)

cv_cnt = 5

for i in seed_range:
    train_res  = []
    res        = []
    test_preds = []

    # Run for each ELM model
    for elm_param in [
        {'n': elm1.n, 'l2_param': elm1.l2_param},
        {'n': elm2.n, 'l2_param': elm2.l2_param}]:

        elm      = ELMclf(random_state=i, **elm_param)
        elm_desc = {'NeuronsCnt': elm.n, 'L2': elm.l2_param}
        elm.fit(x_train.to_numpy(), y_train.to_numpy())

        # Predict Train
        y_pred_train  = elm.predict(x_train.to_numpy())
        train_metrics = elm_desc | get_metrics(y_train.to_numpy(), np.array(y_pred_train))
        train_res.append(train_metrics)

        # Predict Test
        y_pred        = elm.predict(x_test.to_numpy())
        outer_metrics = elm_desc | get_metrics(y_test.to_numpy(), np.array(y_pred))
        res.append(outer_metrics)
        test_preds.append(y_pred)

        # Cross Validation
        train_cv, test_cv = cross_val_score(elm, X.to_numpy(), Y.to_numpy(), scoring=get_metrics, cv=cv_cnt, ret_train=True)

        train_df = pd.DataFrame(train_cv)
        train_df.insert(0, 'Fold',       range(1,cv_cnt+1))
        train_df.insert(0, 'Seed',       i)
        train_df.insert(0, 'NeuronsCnt', elm.n)
        train_df.insert(0, 'L2',         elm.l2_param)
        big_train_multifold.append(train_df)

        test_df = pd.DataFrame(test_cv)
        test_df.insert(0, 'Fold',       range(1,cv_cnt+1))
        test_df.insert(0, 'Seed',       i)
        test_df.insert(0, 'NeuronsCnt', elm.n)
        test_df.insert(0, 'L2',         elm.l2_param)
        big_test_multifold.append(test_df)

    # Append Train DF
    train_metrics_df = pd.DataFrame(train_res)
    train_metrics_df = pd.concat(
        (pd.DataFrame([
            f' ELM with {int(elm_best_neuroncnt) } neurons',
            f'RELM with {int(elm_best_neuroncnt) } neurons, L2: {relm_best_l2}'
        ], columns=['Description']), train_metrics_df
        ), axis=1)
    big_train_res.append(train_metrics_df)

    # Append Test Df
    metrics_df = pd.DataFrame(res)
    metrics_df = pd.concat(
        (pd.DataFrame([
            f' ELM with {int(elm_best_neuroncnt) } neurons',
            f'RELM with {int(elm_best_neuroncnt) } neurons, L2: {relm_best_l2}'
        ], columns=['Description']), metrics_df
        ), axis=1)
    big_res.append(metrics_df)
```

<h3>Place CV Multi Seed to CSV</h3>

```python
big_test_multi_df = pd.concat(big_test_multifold)
elm_multi_df  = big_test_multi_df.loc[
    (big_test_multi_df['L2'] == 0)
].drop(['NeuronsCnt', 'L2'], axis=1)
relm_multi_df = big_test_multi_df.loc[
    (big_test_multi_df['L2'] != 0) &
    (big_test_multi_df['NeuronsCnt'] == big_test_multi_df['NeuronsCnt'].max())
].drop(['NeuronsCnt', 'L2'], axis=1)

relm_sub_multi_df = big_test_multi_df.loc[
    (big_test_multi_df['L2'] != 0) &
    (big_test_multi_df['NeuronsCnt'] == big_test_multi_df['NeuronsCnt'].min())
].drop(['NeuronsCnt', 'L2'], axis=1)
```

```python
elm_cv_fname  = f'{path_to_data_imp}result_cv_elm_n{int(elm_best_neuroncnt)}.csv'
relm_cv_fname = f'{path_to_data_imp}result_cv_relm_n{int(elm_best_neuroncnt)}_l{int(relm_best_l2*100)}.csv'
```

```python
# to csv
elm_multi_df.to_csv(elm_cv_fname)
relm_multi_df.to_csv(relm_cv_fname)
```

<h3>Place Single Test Result, Multi Seed, to CSV</h3>

```python
big_res_df = pd.concat(big_res)
elm_single_df  = big_res_df.loc[
    big_res_df['L2'] == 0
].drop(['NeuronsCnt', 'L2', 'Description'], axis=1)
relm_single_df = big_res_df.loc[
    (big_res_df['L2'] != 0) &
    (big_res_df['NeuronsCnt'] == big_res_df['NeuronsCnt'].max())
].drop(['NeuronsCnt', 'L2', 'Description'], axis=1)
```

```python
elm_fname  = f'{path_to_data_imp}result_elm_n{int(elm_best_neuroncnt)}.csv'
relm_fname = f'{path_to_data_imp}result_relm_n{int(elm_best_neuroncnt)}_l{int(relm_best_l2*100)}.csv'
```

```python
# to csv
elm_single_df.to_csv(elm_fname)
relm_single_df.to_csv(relm_fname)
```
